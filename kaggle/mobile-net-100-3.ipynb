```python

# ================================================================
# JFAMP — STAGE 4B: SERIALIZED PAYLOAD ROUND-TRIP VALIDATION
# ================================================================

import json
import hashlib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from pathlib import Path
from torchvision.models import mobilenet_v2

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
CHECKPOINT_PATH = ROOT / "checkpoints/mobilenetv2_cifar100_best.pth"
STAGE2 = ROOT / "experiments/stage_02_baseline"
STAGE4 = ROOT / "experiments/stage_04_fault_injection"

REFERENCE_CSV = STAGE2 / "codec_validation_per_layer.csv"
REFERENCE_JSON = STAGE2 / "codec_validation_summary.json"

FORMATS = [
    "FP16", "BF16", "E4M11_SN",
    "FP8_E4M3FN", "FP8_E5M2", "INT8", "INT4"
]

print("=" * 72)
print("JFAMP — STAGE 4B: SERIALIZED PAYLOAD VALIDATION")
print("=" * 72)

# ------------------------------------------------
# 1. Required artifact checks
# ------------------------------------------------

for path in [CHECKPOINT_PATH, REFERENCE_CSV, REFERENCE_JSON]:
    if not path.is_file():
        raise FileNotFoundError(f"Required artifact missing: {path}")
    print(f"[OK] {path.name}: {path.stat().st_size:,} bytes")

# ------------------------------------------------
# 2. E4M11-SN encoder and decoder
# ------------------------------------------------

FRAC_BITS = 11
BIAS = 7
MIN_NORMAL = 2.0 ** (1 - BIAS)
MIN_SUBNORMAL = 2.0 ** (1 - BIAS - FRAC_BITS)

def encode_e4m11(values):
    x = np.asarray(values, dtype=np.float64)
    flat = x.reshape(-1)
    encoded = np.zeros(flat.shape, dtype=np.uint16)

    for i, value in enumerate(flat):
        sign = 1 if np.signbit(value) else 0
        sign_bits = sign << 15
        magnitude = abs(value)

        if np.isnan(value):
            encoded[i] = sign_bits | (15 << FRAC_BITS) | 1
        elif np.isinf(value):
            encoded[i] = sign_bits | (15 << FRAC_BITS)
        elif magnitude == 0:
            encoded[i] = sign_bits
        elif magnitude < MIN_NORMAL:
            fraction = int(np.rint(magnitude / MIN_SUBNORMAL))
            if fraction == 0:
                encoded[i] = sign_bits
            elif fraction >= (1 << FRAC_BITS):
                encoded[i] = sign_bits | (1 << FRAC_BITS)
            else:
                encoded[i] = sign_bits | fraction
        else:
            exponent = int(np.floor(np.log2(magnitude)))
            exponent_field = exponent + BIAS
            significand = magnitude / (2.0 ** exponent)

            fraction = int(
                np.rint((significand - 1.0) * (1 << FRAC_BITS))
            )

            if fraction >= (1 << FRAC_BITS):
                fraction = 0
                exponent_field += 1

            if exponent_field >= 15:
                encoded[i] = sign_bits | (15 << FRAC_BITS)
            else:
                encoded[i] = (
                    sign_bits
                    | (exponent_field << FRAC_BITS)
                    | fraction
                )

    return encoded.reshape(x.shape)


def decode_e4m11(encoded):
    bits = np.asarray(encoded, dtype=np.uint16)
    flat = bits.reshape(-1)
    decoded = np.zeros(flat.shape, dtype=np.float64)

    for i, word in enumerate(flat):
        sign = -1.0 if (int(word) >> 15) & 1 else 1.0
        exponent_field = (int(word) >> FRAC_BITS) & 0xF
        fraction = int(word) & 0x7FF

        if exponent_field == 15:
            decoded[i] = (
                sign * np.inf if fraction == 0 else np.nan
            )
        elif exponent_field == 0:
            decoded[i] = fraction * MIN_SUBNORMAL * sign
        else:
            exponent = exponent_field - BIAS
            significand = 1.0 + fraction / (1 << FRAC_BITS)
            decoded[i] = sign * significand * (2.0 ** exponent)

    return decoded.reshape(bits.shape)

# ------------------------------------------------
# 3. INT4 packing and unpacking
# ------------------------------------------------

def pack_int4(values):
    values = np.asarray(values, dtype=np.int8)
    original_length = len(values)

    assert np.all(values >= -7) and np.all(values <= 7)

    padded = values.copy()
    if len(padded) % 2:
        padded = np.append(padded, np.int8(0))

    unsigned = (padded.astype(np.int16) & 0x0F).astype(np.uint8)

    packed = (
        unsigned[0::2] | (unsigned[1::2] << 4)
    ).astype(np.uint8)

    return packed, original_length


def unpack_int4(packed, original_length):
    packed = np.asarray(packed, dtype=np.uint8)

    low = packed & 0x0F
    high = (packed >> 4) & 0x0F

    values = np.empty(len(packed) * 2, dtype=np.int8)

    values[0::2] = np.where(
        low >= 8, low.astype(np.int16) - 16, low
    )
    values[1::2] = np.where(
        high >= 8, high.astype(np.int16) - 16, high
    )

    return values[:original_length]

# ------------------------------------------------
# 4. Unified serialized payload encoder
# ------------------------------------------------

def encode_weights(weights, fmt):
    x = np.asarray(weights, dtype=np.float32).reshape(-1)

    if fmt == "FP16":
        t = torch.from_numpy(x.copy()).to(torch.float16)
        payload = t.view(torch.uint8).numpy().copy()
        metadata = {"shape": list(weights.shape)}

    elif fmt == "BF16":
        t = torch.from_numpy(x.copy()).to(torch.bfloat16)
        payload = t.view(torch.uint8).numpy().copy()
        metadata = {"shape": list(weights.shape)}

    elif fmt == "FP8_E4M3FN":
        t = torch.from_numpy(x.copy()).to(torch.float8_e4m3fn)
        payload = t.view(torch.uint8).numpy().copy()
        metadata = {"shape": list(weights.shape)}

    elif fmt == "FP8_E5M2":
        t = torch.from_numpy(x.copy()).to(torch.float8_e5m2)
        payload = t.view(torch.uint8).numpy().copy()
        metadata = {"shape": list(weights.shape)}

    elif fmt == "E4M11_SN":
        encoded = encode_e4m11(x)
        payload = encoded.view(np.uint8).copy()
        metadata = {
            "shape": list(weights.shape),
            "element_count": len(x)
        }

    elif fmt in ("INT8", "INT4"):
        bits = 8 if fmt == "INT8" else 4
        qmax = 127 if bits == 8 else 7

        max_abs = float(np.max(np.abs(x)))
        scale = max_abs / qmax if max_abs > 0 else 1.0

        q = np.rint(x / scale)
        q = np.clip(q, -qmax, qmax).astype(np.int8)

        if fmt == "INT8":
            payload = q.view(np.uint8).copy()
        else:
            payload, _ = pack_int4(q)

        metadata = {
            "shape": list(weights.shape),
            "element_count": len(x),
            "scale": float(scale),
            "qmin": -qmax,
            "qmax": qmax,
            "packing": "low_nibble_first" if bits == 4 else None
        }

    else:
        raise ValueError(f"Unsupported format: {fmt}")

    return payload, metadata


def decode_weights(payload, metadata, fmt):
    shape = tuple(metadata["shape"])
    count = metadata.get("element_count", int(np.prod(shape)))

    if fmt == "FP16":
        decoded = torch.from_numpy(
            payload.copy()
        ).view(torch.float16).float().numpy()

    elif fmt == "BF16":
        decoded = torch.from_numpy(
            payload.copy()
        ).view(torch.bfloat16).float().numpy()

    elif fmt == "FP8_E4M3FN":
        decoded = torch.from_numpy(
            payload.copy()
        ).view(torch.float8_e4m3fn).float().numpy()

    elif fmt == "FP8_E5M2":
        decoded = torch.from_numpy(
            payload.copy()
        ).view(torch.float8_e5m2).float().numpy()

    elif fmt == "E4M11_SN":
        encoded = payload.view(np.uint16)
        decoded = decode_e4m11(encoded).astype(np.float32)

    elif fmt == "INT8":
        q = payload.view(np.int8)
        decoded = q.astype(np.float32) * metadata["scale"]

    elif fmt == "INT4":
        q = unpack_int4(payload, count)
        decoded = q.astype(np.float32) * metadata["scale"]

    else:
        raise ValueError(f"Unsupported format: {fmt}")

    return decoded[:count].reshape(shape)

# ------------------------------------------------
# 5. Load checkpoint and model
# ------------------------------------------------

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False
)

model = mobilenet_v2(weights=None, num_classes=100)
model.features[0][0].stride = (1, 1)
model.load_state_dict(checkpoint["model_state_dict"], strict=True)
model.eval()

weight_tensors = {
    name: module.weight.detach().cpu().float().numpy().copy()
    for name, module in model.named_modules()
    if isinstance(module, (nn.Conv2d, nn.Linear))
}

assert len(weight_tensors) == 53
assert sum(x.size for x in weight_tensors.values()) == 2_317_760

print("\n[OK] Model reconstructed")
print("[OK] Quantizable layers:", len(weight_tensors))
print("[OK] Quantizable weights:", f"{sum(x.size for x in weight_tensors.values()):,}")

# ------------------------------------------------
# 6. Run all 371 layer-format validations
# ------------------------------------------------

reference = pd.read_csv(REFERENCE_CSV)
reference_summary = pd.DataFrame(
    json.loads(REFERENCE_JSON.read_text())
)

records = []
payload_hashes = {}

for fmt in FORMATS:
    print(f"\nValidating {fmt} ...")

    for layer_name, original in weight_tensors.items():
        payload, metadata = encode_weights(original, fmt)
        reconstructed = decode_weights(payload, metadata, fmt)

        assert reconstructed.shape == original.shape
        assert np.isfinite(reconstructed).all()

        error = (
            reconstructed.astype(np.float64)
            - original.astype(np.float64)
        )

        records.append({
            "format": fmt,
            "layer": layer_name,
            "weights": int(original.size),
            "payload_bytes": int(payload.nbytes),
            "mse": float(np.mean(error ** 2)),
            "max_abs_error": float(np.max(np.abs(error))),
            "nonfinite_count": int(
                np.count_nonzero(~np.isfinite(reconstructed))
            ),
            "roundtrip_shape_valid": True
        })

        payload_hashes[f"{fmt}::{layer_name}"] = hashlib.sha256(
            payload.tobytes()
        ).hexdigest()

results = pd.DataFrame(records)

assert len(results) == 371
assert results.groupby("format").size().eq(53).all()

# ------------------------------------------------
# 7. Compare against frozen Stage 2 per-layer results
# ------------------------------------------------

merged = results.merge(
    reference,
    on=["format", "layer"],
    suffixes=("_new", "_reference"),
    validate="one_to_one"
)

assert len(merged) == 371

assert np.array_equal(
    merged["payload_bytes_new"].to_numpy(),
    merged["payload_bytes_reference"].to_numpy()
), "Payload byte counts differ from Stage 2."

mse_diff = np.abs(
    merged["mse_new"].to_numpy()
    - merged["mse_reference"].to_numpy()
)

max_error_diff = np.abs(
    merged["max_abs_error_new"].to_numpy()
    - merged["max_abs_error_reference"].to_numpy()
)

print("\nPER-LAYER REFERENCE COMPARISON")
print("Matched records:", len(merged))
print("Maximum MSE difference:", float(mse_diff.max()))
print("Maximum max-error difference:", float(max_error_diff.max()))

assert np.allclose(
    merged["mse_new"],
    merged["mse_reference"],
    rtol=1e-6,
    atol=1e-12
), "Per-layer MSE differs from Stage 2."

assert np.allclose(
    merged["max_abs_error_new"],
    merged["max_abs_error_reference"],
    rtol=1e-6,
    atol=1e-12
), "Per-layer maximum error differs from Stage 2."

# ------------------------------------------------
# 8. Aggregate payload-size comparison
# ------------------------------------------------

aggregate = (
    results.groupby("format", as_index=False)
    .agg(
        layers=("layer", "count"),
        total_payload_bytes=("payload_bytes", "sum"),
        nonfinite_values=("nonfinite_count", "sum")
    )
)

summary_merge = aggregate.merge(
    reference_summary[["format", "payload_bytes"]],
    on="format",
    suffixes=("_new", "_reference"),
    validate="one_to_one"
)

assert (
    summary_merge["total_payload_bytes_new"]
    == summary_merge["payload_bytes_reference"]
).all()

assert (aggregate["nonfinite_values"] == 0).all()

print("\nAGGREGATE PAYLOAD AUDIT")
print(aggregate.to_string(index=False))

# ------------------------------------------------
# 9. Save Stage 4B artifacts
# ------------------------------------------------

STAGE4.mkdir(parents=True, exist_ok=True)
(STAGE4 / "manifests").mkdir(parents=True, exist_ok=True)
(STAGE4 / "results").mkdir(parents=True, exist_ok=True)

results_path = (
    STAGE4 / "results" / "stage4b_clean_payload_roundtrip.csv"
)
hash_path = (
    STAGE4 / "manifests" / "stage4b_payload_hashes.json"
)
audit_path = (
    STAGE4 / "manifests" / "stage4b_codec_audit.json"
)

results.to_csv(results_path, index=False)

with open(hash_path, "w") as f:
    json.dump(payload_hashes, f, indent=2)

audit = {
    "stage": "04B_clean_serialized_payload_roundtrip",
    "checkpoint": CHECKPOINT_PATH.name,
    "checkpoint_sha256": hashlib.sha256(
        CHECKPOINT_PATH.read_bytes()
    ).hexdigest(),
    "codec_source": "Original unified Stage 2 Cell 12 implementation",
    "formats": FORMATS,
    "layers": len(weight_tensors),
    "quantizable_weights": int(
        sum(x.size for x in weight_tensors.values())
    ),
    "validated_layer_format_pairs": len(results),
    "reference_pairs_matched": len(merged),
    "payload_sizes_match_stage2": True,
    "per_layer_mse_match_stage2": True,
    "per_layer_max_error_match_stage2": True,
    "all_reconstructed_weights_finite": True,
    "payload_hashes_saved": len(payload_hashes),
    "faults_injected": 0,
    "status": "passed"
}

with open(audit_path, "w") as f:
    json.dump(audit, f, indent=2)

print("\n" + "=" * 72)
print("STAGE 4B RESULT: PASSED")
print("=" * 72)
print("Validated combinations:", len(results))
print("Payload hashes saved:", len(payload_hashes))
print("Faults injected: 0")
print("Results:", results_path)
print("Hashes:", hash_path)
print("Audit:", audit_path)
```

```python

# ================================================================
# JFAMP — STAGE 4C.1: ENCODED PAYLOAD BIT-FLIP SMOKE TEST
# Tests all 7 formats with 1-bit and 8-bit injections.
# Does not modify clean payloads or run model inference.
# ================================================================

import json
import hashlib
import numpy as np
import pandas as pd
from pathlib import Path

SEED = 20261114
FAULT_GROUPS = [1, 8]

assert len(weight_tensors) == 53
assert callable(encode_weights)
assert callable(decode_weights)

# Use the first inventory layer as a representative test layer.
TEST_LAYER = sorted(weight_tensors.keys())[0]
TEST_WEIGHTS = weight_tensors[TEST_LAYER]

print("=" * 72)
print("JFAMP — STAGE 4C.1: BIT-FLIP INJECTION SMOKE TEST")
print("=" * 72)
print("Representative layer:", TEST_LAYER)
print("Weight shape:", TEST_WEIGHTS.shape)
print("Fault groups:", FAULT_GROUPS)
print("Random seed:", SEED)

# ------------------------------------------------
# 1. Create isolated output directory
# ------------------------------------------------

SMOKE_DIR = (
    STAGE4 / "fault_trials" / "stage4c_smoke_test"
)
SMOKE_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------
# 2. Deterministic bit-flip function
# ------------------------------------------------

def inject_bit_flips(clean_payload, number_of_flips, seed):
    """
    Flip exactly number_of_flips unique bits in a serialized
    uint8 payload. Bit position p maps to:
        byte index = p // 8
        bit index  = p % 8
    """

    clean = np.asarray(clean_payload, dtype=np.uint8)

    if clean.size == 0:
        raise ValueError("Cannot inject faults into an empty payload.")

    total_bits = clean.size * 8

    if number_of_flips > total_bits:
        raise ValueError("Requested flips exceed available payload bits.")

    rng = np.random.default_rng(seed)

    selected_positions = rng.choice(
        total_bits,
        size=number_of_flips,
        replace=False
    )

    corrupted = clean.copy()

    for position in selected_positions:
        byte_index = int(position // 8)
        bit_index = int(position % 8)
        corrupted[byte_index] ^= np.uint8(1 << bit_index)

    actual_changed_bits = int(
        np.unpackbits(
            np.bitwise_xor(clean, corrupted)
        ).sum()
    )

    assert actual_changed_bits == number_of_flips
    assert not np.array_equal(clean, corrupted)

    return corrupted, selected_positions.tolist()


# ------------------------------------------------
# 3. Test all formats and both fault groups
# ------------------------------------------------

records = []
clean_payload_hashes = {}
corrupted_payload_hashes = {}

for fmt_index, fmt in enumerate(FORMATS):

    clean_payload, metadata = encode_weights(
        TEST_WEIGHTS, fmt
    )

    clean_payload = np.asarray(
        clean_payload, dtype=np.uint8
    ).copy()

    clean_hash = hashlib.sha256(
        clean_payload.tobytes()
    ).hexdigest()

    clean_payload_hashes[fmt] = clean_hash

    # Preserve an independent immutable reference.
    clean_reference = clean_payload.copy()

    for fault_group in FAULT_GROUPS:

        # Derive deterministic but independent seeds.
        trial_seed = (
            SEED
            + fmt_index * 1000
            + fault_group
        )

        corrupted_payload, bit_positions = inject_bit_flips(
            clean_payload,
            fault_group,
            trial_seed
        )

        # Verify clean reference was not modified.
        assert np.array_equal(
            clean_payload, clean_reference
        )

        # Verify exact changed-bit count.
        xor_payload = np.bitwise_xor(
            clean_payload, corrupted_payload
        )

        changed_bits = int(
            np.unpackbits(xor_payload).sum()
        )

        assert changed_bits == fault_group

        corrupted_hash = hashlib.sha256(
            corrupted_payload.tobytes()
        ).hexdigest()

        assert corrupted_hash != clean_hash

        # Decode the actual corrupted serialized payload.
        reconstructed = decode_weights(
            corrupted_payload,
            metadata,
            fmt
        )

        assert reconstructed.shape == TEST_WEIGHTS.shape

        # Non-finite values are recorded, not silently discarded.
        nonfinite_count = int(
            np.count_nonzero(~np.isfinite(reconstructed))
        )

        records.append({
            "layer": TEST_LAYER,
            "format": fmt,
            "fault_group_bits": fault_group,
            "seed": trial_seed,
            "payload_bytes": int(clean_payload.nbytes),
            "total_payload_bits": int(clean_payload.nbytes * 8),
            "actual_changed_bits": changed_bits,
            "selected_bit_positions": json.dumps(bit_positions),
            "clean_payload_sha256": clean_hash,
            "corrupted_payload_sha256": corrupted_hash,
            "reconstructed_shape_valid": True,
            "nonfinite_reconstructed_values": nonfinite_count
        })

        # Save corrupted payload separately.
        payload_path = (
            SMOKE_DIR
            / f"{fmt}_{fault_group}bit_corrupted_payload.bin"
        )

        corrupted_payload.tofile(payload_path)

        print(
            f"[PASS] {fmt:12s} | "
            f"{fault_group:2d} bit flips | "
            f"changed={changed_bits:2d} | "
            f"nonfinite={nonfinite_count}"
        )

        # Confirm clean payload is still unchanged.
        assert hashlib.sha256(
            clean_payload.tobytes()
        ).hexdigest() == clean_hash


# ------------------------------------------------
# 4. Save smoke-test records and protocol manifest
# ------------------------------------------------

smoke_results = pd.DataFrame(records)

assert len(smoke_results) == 14
assert smoke_results["actual_changed_bits"].equals(
    smoke_results["fault_group_bits"]
)
assert smoke_results["reconstructed_shape_valid"].all()

results_path = SMOKE_DIR / "stage4c_smoke_test_results.csv"
manifest_path = SMOKE_DIR / "stage4c_smoke_test_manifest.json"

smoke_results.to_csv(results_path, index=False)

manifest = {
    "stage": "04C_encoded_payload_bit_flip_injection",
    "test_type": "smoke_test",
    "representative_layer": TEST_LAYER,
    "formats": FORMATS,
    "fault_groups_bits": FAULT_GROUPS,
    "random_seed": SEED,
    "bit_selection": "uniform random unique positions without replacement",
    "bit_numbering": "LSB-first within each serialized byte",
    "injection_target": "serialized quantized weight payload bytes",
    "injection_location": "after encoding and before decoding",
    "clean_payloads_modified": False,
    "tested_combinations": len(smoke_results),
    "exact_flip_counts_verified": True,
    "decode_shape_verified": True,
    "model_inference_performed": False,
    "status": "passed"
}

with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

print("\n" + "=" * 72)
print("STAGE 4C.1 RESULT: PASSED")
print("=" * 72)
print("Representative layer:", TEST_LAYER)
print("Formats tested:", len(FORMATS))
print("Fault groups tested:", len(FAULT_GROUPS))
print("Total smoke tests:", len(smoke_results))
print("Exact bit flips verified: 14/14")
print("Clean payloads modified: No")
print("Model inference performed: No")
print("Results:", results_path)
print("Manifest:", manifest_path)
```

```python

# ================================================================
# JFAMP — STAGE 4C.2: FULL-LAYER ENCODED FAULT CHARACTERIZATION
# 53 layers × 7 formats × 2 fault groups = 742 trials
# No model inference. No changes to the clean checkpoint.
# ================================================================

import json
import hashlib
import numpy as np
import pandas as pd
from pathlib import Path

SEED = 20261114
FAULT_GROUPS = [1, 8]

assert len(weight_tensors) == 53
assert len(FORMATS) == 7
assert callable(inject_bit_flips)

OUT_DIR = STAGE4 / "fault_trials" / "stage4c_full_layer"
OUT_DIR.mkdir(parents=True, exist_ok=True)

records = []

# Use sorted names for reproducible iteration.
layer_names = sorted(weight_tensors)

for fmt_idx, fmt in enumerate(FORMATS):
    print(f"\nProcessing {fmt} ...")

    for layer_idx, layer_name in enumerate(layer_names):
        original = weight_tensors[layer_name]

        clean_payload, metadata = encode_weights(original, fmt)
        clean_payload = np.asarray(
            clean_payload, dtype=np.uint8
        ).copy()

        clean_hash = hashlib.sha256(
            clean_payload.tobytes()
        ).hexdigest()

        clean_decoded = decode_weights(
            clean_payload, metadata, fmt
        )

        for fault_group in FAULT_GROUPS:
            trial_seed = (
                SEED
                + fmt_idx * 100000
                + layer_idx * 100
                + fault_group
            )

            corrupted, positions = inject_bit_flips(
                clean_payload, fault_group, trial_seed
            )

            changed_bits = int(
                np.unpackbits(
                    np.bitwise_xor(clean_payload, corrupted)
                ).sum()
            )
            assert changed_bits == fault_group

            decoded = decode_weights(corrupted, metadata, fmt)
            assert decoded.shape == original.shape

            nonfinite_mask = ~np.isfinite(decoded)
            nonfinite_count = int(nonfinite_mask.sum())

            # Compare only finite entries; preserve nonfinite count.
            finite_mask = np.isfinite(decoded) & np.isfinite(clean_decoded)
            finite_errors = (
                decoded[finite_mask].astype(np.float64)
                - clean_decoded[finite_mask].astype(np.float64)
            )

            mse = (
                float(np.mean(finite_errors ** 2))
                if finite_errors.size else None
            )
            max_error = (
                float(np.max(np.abs(finite_errors)))
                if finite_errors.size else None
            )

            records.append({
                "format": fmt,
                "layer": layer_name,
                "fault_group_bits": fault_group,
                "seed": trial_seed,
                "weight_count": int(original.size),
                "payload_bytes": int(clean_payload.size),
                "changed_bits": changed_bits,
                "selected_bit_positions": json.dumps(positions),
                "nonfinite_values": nonfinite_count,
                "finite_weight_mse": mse,
                "finite_max_abs_error": max_error,
                "clean_payload_sha256": clean_hash,
                "corrupted_payload_sha256": hashlib.sha256(
                    corrupted.tobytes()
                ).hexdigest()
            })

        # Ensure original payload was never mutated.
        assert hashlib.sha256(
            clean_payload.tobytes()
        ).hexdigest() == clean_hash

    print(f"[DONE] {fmt}: {len(layer_names) * 2} trials")

df = pd.DataFrame(records)

assert len(df) == 742
assert (df["changed_bits"] == df["fault_group_bits"]).all()
assert not df.duplicated(
    ["format", "layer", "fault_group_bits"]
).any()

results_path = OUT_DIR / "stage4c_full_layer_fault_characterization.csv"
df.to_csv(results_path, index=False)

summary = (
    df.groupby(["format", "fault_group_bits"], as_index=False)
    .agg(
        trials=("layer", "count"),
        trials_with_nonfinite=("nonfinite_values", lambda x: int((x > 0).sum())),
        total_nonfinite_values=("nonfinite_values", "sum"),
        mean_finite_weight_mse=("finite_weight_mse", "mean"),
        max_finite_abs_error=("finite_max_abs_error", "max")
    )
)

summary_path = OUT_DIR / "stage4c_fault_summary.csv"
summary.to_csv(summary_path, index=False)

manifest = {
    "stage": "04C_full_layer_fault_characterization",
    "seed": SEED,
    "formats": FORMATS,
    "layers": 53,
    "fault_groups_bits": FAULT_GROUPS,
    "trials_per_combination": 1,
    "total_trials": 742,
    "injection_target": "serialized quantized weight payload",
    "bit_selection": "uniform unique positions without replacement",
    "fault_scope": "one layer per trial",
    "comparison_reference": "clean decoded quantized weights",
    "nonfinite_policy": "record separately; exclude from finite-error metrics",
    "model_inference_performed": False,
    "status": "completed"
}

manifest_path = OUT_DIR / "stage4c_full_layer_manifest.json"
manifest_path.write_text(json.dumps(manifest, indent=2))

print("\n" + "=" * 72)
print("STAGE 4C.2 RESULT")
print("=" * 72)
print("Total trials:", len(df))
print("Exact bit-count checks: PASSED")
print("Clean payload integrity: PASSED")
print("\nSUMMARY:")
print(summary.to_string(index=False))
print("\nResults:", results_path)
print("Summary:", summary_path)
print("Manifest:", manifest_path)
```

```python

# ================================================================
# JFAMP — STAGE 4C.3: MODEL-LEVEL INFERENCE SMOKE TEST
# One layer, one format, one actual serialized payload bit flip.
# ================================================================

import copy
import numpy as np
import torch
import torch.nn as nn

SEED = 20261114
TEST_LAYER = "classifier.1"
TEST_FORMAT = "FP16"
FAULT_BITS = 1

assert TEST_LAYER in weight_tensors
assert "val_loader" in globals(), (
    "val_loader is not available. Stop and share this message."
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ------------------------------------------------
# 1. Helper: evaluate a model on the same validation loader
# ------------------------------------------------

def evaluate_model(model_to_test):
    model_to_test.eval()
    model_to_test.to(device)

    correct = 0
    total = 0
    loss_sum = 0.0
    predictions = []
    labels_all = []

    criterion = nn.CrossEntropyLoss(reduction="sum")

    with torch.inference_mode():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            logits = model_to_test(images)
            loss_sum += criterion(logits, labels).item()

            preds = logits.argmax(dim=1)

            correct += (preds == labels).sum().item()
            total += labels.numel()

            predictions.append(preds.cpu().numpy())
            labels_all.append(labels.cpu().numpy())

    model_to_test.cpu()

    predictions = np.concatenate(predictions)
    labels_all = np.concatenate(labels_all)

    return {
        "accuracy": 100.0 * correct / total,
        "loss": loss_sum / total,
        "correct": correct,
        "total": total,
        "predictions": predictions,
        "labels": labels_all
    }


# ------------------------------------------------
# 2. FP32 baseline
# ------------------------------------------------

baseline_model = copy.deepcopy(model).cpu()
baseline_result = evaluate_model(baseline_model)

print("FP32 baseline accuracy:", baseline_result["accuracy"])
print("FP32 baseline loss:", baseline_result["loss"])


# ------------------------------------------------
# 3. Serialize the selected layer's weights
# ------------------------------------------------

original_weights = weight_tensors[TEST_LAYER]

clean_payload, metadata = encode_weights(
    original_weights, TEST_FORMAT
)

clean_payload = np.asarray(
    clean_payload, dtype=np.uint8
).copy()

clean_decoded = decode_weights(
    clean_payload, metadata, TEST_FORMAT
)

# ------------------------------------------------
# 4. Create clean one-layer quantized model
# ------------------------------------------------

clean_quant_model = copy.deepcopy(model).cpu()

module_map = dict(clean_quant_model.named_modules())
target_module = module_map[TEST_LAYER]

target_module.weight.data.copy_(
    torch.from_numpy(clean_decoded.copy()).to(
        dtype=target_module.weight.dtype
    )
)

clean_quant_result = evaluate_model(clean_quant_model)

print("\nClean one-layer quantized accuracy:",
      clean_quant_result["accuracy"])

# ------------------------------------------------
# 5. Inject one bit into the serialized payload
# ------------------------------------------------

corrupted_payload, positions = inject_bit_flips(
    clean_payload,
    FAULT_BITS,
    SEED + 123
)

actual_flips = int(
    np.unpackbits(
        np.bitwise_xor(clean_payload, corrupted_payload)
    ).sum()
)

assert actual_flips == FAULT_BITS
assert not np.array_equal(clean_payload, corrupted_payload)

corrupted_decoded = decode_weights(
    corrupted_payload, metadata, TEST_FORMAT
)

assert corrupted_decoded.shape == original_weights.shape

print("\nInjected bit positions:", positions)
print("Actual changed bits:", actual_flips)
print("Non-finite decoded values:",
      int((~np.isfinite(corrupted_decoded)).sum()))

# ------------------------------------------------
# 6. Create corrupted one-layer model
# ------------------------------------------------

corrupted_model = copy.deepcopy(model).cpu()

corrupted_module = dict(
    corrupted_model.named_modules()
)[TEST_LAYER]

corrupted_module.weight.data.copy_(
    torch.from_numpy(corrupted_decoded.copy()).to(
        dtype=corrupted_module.weight.dtype
    )
)

corrupted_result = evaluate_model(corrupted_model)

# ------------------------------------------------
# 7. Compare predictions and accuracy
# ------------------------------------------------

clean_preds = clean_quant_result["predictions"]
fault_preds = corrupted_result["predictions"]
base_preds = baseline_result["predictions"]

assert np.array_equal(
    clean_quant_result["labels"],
    corrupted_result["labels"]
)

changed_vs_clean = int(np.sum(fault_preds != clean_preds))
changed_vs_fp32 = int(np.sum(fault_preds != base_preds))

clean_correct = clean_preds == baseline_result["labels"]
fault_correct = fault_preds == baseline_result["labels"]

correct_to_wrong = int(np.sum(clean_correct & ~fault_correct))
wrong_to_correct = int(np.sum(~clean_correct & fault_correct))

print("\n" + "=" * 72)
print("STAGE 4C.3 — MODEL-LEVEL RESULTS")
print("=" * 72)

print(f"FP32 baseline accuracy:          {baseline_result['accuracy']:.4f}%")
print(f"Clean quantized accuracy:        {clean_quant_result['accuracy']:.4f}%")
print(f"Corrupted payload accuracy:      {corrupted_result['accuracy']:.4f}%")

print(f"\nClean quantization delta:        "
      f"{clean_quant_result['accuracy'] - baseline_result['accuracy']:+.4f} pp")

print(f"Fault delta vs clean quantized:  "
      f"{corrupted_result['accuracy'] - clean_quant_result['accuracy']:+.4f} pp")

print(f"Fault delta vs FP32:             "
      f"{corrupted_result['accuracy'] - baseline_result['accuracy']:+.4f} pp")

print("\nPrediction changes vs clean quantized:", changed_vs_clean)
print("Prediction changes vs FP32:", changed_vs_fp32)
print("Correct -> wrong:", correct_to_wrong)
print("Wrong -> correct:", wrong_to_correct)

print("\n[PASS] Serialized payload was corrupted before decoding.")
print("[PASS] Corrupted weights were loaded into the model.")
print("[PASS] All three model evaluations completed.")
print("Device:", device)
```

```python

# =====================================================================
# JFAMP — STAGE 4C.3: MODEL-LEVEL ENCODED PAYLOAD FAULT SMOKE TEST
#
# Three evaluations:
#   A. Original FP32 baseline
#   B. Clean FP16 encoding/decoding of classifier.1 weights
#   C. One-bit corruption of the serialized FP16 payload
#
# Fault is injected into the actual encoded payload BEFORE decoding.
# Only classifier.1 weights are quantized/corrupted; all other model
# parameters and biases remain FP32.
# =====================================================================

import copy
import json
import hashlib
import inspect
import time
import numpy as np
import pandas as pd
import torch
from pathlib import Path

# ----------------------------- CONFIG -------------------------------

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
OUT = ROOT / "experiments/stage_04_fault_injection/results/stage4c3"
OUT.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PATH = ROOT / "checkpoints/mobilenetv2_cifar100_best.pth"
REFERENCE_PATH = (
    ROOT / "experiments/stage_02_baseline/"
    "fp32_reference_predictions.npz"
)

LAYER_NAME = "classifier.1"
FORMAT_NAME = "FP16"
FAULT_BITS = 1
SEED = 20261114

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# -------------------------- INPUT VALIDATION -------------------------

assert CHECKPOINT_PATH.is_file(), "Checkpoint missing."
assert REFERENCE_PATH.is_file(), "Reference predictions missing."
assert "val_loader" in globals(), "Validation loader missing."
assert "model" in globals(), "Existing MobileNetV2 model missing."
assert callable(encode_weights), "encode_weights unavailable."
assert callable(decode_weights), "decode_weights unavailable."

reference = np.load(REFERENCE_PATH)
ref_predictions = reference["predictions"].astype(np.int64)
ref_labels = reference["labels"].astype(np.int64)

assert len(ref_predictions) == 5000
assert len(ref_labels) == 5000

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False
)
assert "model_state_dict" in checkpoint

# Reuse the existing architecture and load the canonical checkpoint.
base_model = copy.deepcopy(model).cpu()
base_model.load_state_dict(checkpoint["model_state_dict"], strict=True)
base_model.eval()

named_parameters = dict(base_model.named_parameters())
assert LAYER_NAME + ".weight" in named_parameters, (
    f"{LAYER_NAME}.weight not found."
)

original_weight = named_parameters[LAYER_NAME + ".weight"].detach().cpu().clone()
original_shape = tuple(original_weight.shape)

# -------------------------- CODEC ADAPTER ----------------------------
# Adapt calls to the exact callable signatures already loaded in the
# notebook. No alternative quantization implementation is introduced.

def invoke_codec(fn, payload_or_tensor, fmt, shape=None):
    sig = inspect.signature(fn)
    params = list(sig.parameters.values())

    values = {}
    for p in params:
        n = p.name.lower()

        if any(k in n for k in ("format", "fmt", "precision", "dtype")):
            values[p.name] = fmt
        elif any(k in n for k in ("shape", "original_shape", "weight_shape")):
            if shape is not None:
                values[p.name] = shape
        elif any(k in n for k in ("device",)):
            values[p.name] = "cpu"
        elif any(k in n for k in (
            "weight", "tensor", "value", "data", "payload",
            "encoded", "buffer", "codes", "array"
        )):
            values[p.name] = payload_or_tensor

    required = [
        p for p in params
        if p.default is inspect.Parameter.empty
        and p.kind in (
            inspect.Parameter.POSITIONAL_OR_KEYWORD,
            inspect.Parameter.KEYWORD_ONLY
        )
    ]

    if all(p.name in values for p in required):
        return fn(**values)

    # Common Stage 4B calling conventions, attempted only if needed.
    attempts = (
        ((payload_or_tensor, fmt, shape), {}),
        ((payload_or_tensor, fmt), {}),
        ((payload_or_tensor, shape, fmt), {}),
        ((payload_or_tensor,), {"format_name": fmt, "shape": shape}),
        ((payload_or_tensor,), {"fmt": fmt, "shape": shape}),
        ((payload_or_tensor,), {"format": fmt, "original_shape": shape}),
        ((payload_or_tensor,), {"format_name": fmt}),
        ((payload_or_tensor,), {"fmt": fmt}),
    )

    errors = []
    for args, kwargs in attempts:
        args = tuple(x for x in args if x is not None)
        try:
            return fn(*args, **kwargs)
        except (TypeError, ValueError) as exc:
            errors.append(str(exc))

    raise RuntimeError(
        f"Could not call {fn.__name__}{sig}. "
        f"Last errors: {errors[-3:]}"
    )

def to_numpy_payload(encoded):
    if isinstance(encoded, torch.Tensor):
        encoded = encoded.detach().cpu().numpy()
    if isinstance(encoded, np.ndarray):
        return np.ascontiguousarray(encoded).view(np.uint8).reshape(-1).copy()
    if isinstance(encoded, (bytes, bytearray, memoryview)):
        return np.frombuffer(bytes(encoded), dtype=np.uint8).copy()
    raise TypeError(
        f"Unsupported encoded payload type: {type(encoded)}"
    )

def decode_payload(payload_bytes, shape):
    # Preserve the codec's expected payload representation where possible.
    payload_arg = np.ascontiguousarray(payload_bytes)
    decoded = invoke_codec(
        decode_weights, payload_arg, FORMAT_NAME, shape
    )

    if isinstance(decoded, tuple):
        # Some codecs return (decoded_tensor, metadata).
        decoded = decoded[0]

    if isinstance(decoded, np.ndarray):
        decoded = torch.from_numpy(decoded.copy())
    elif not isinstance(decoded, torch.Tensor):
        decoded = torch.as_tensor(decoded)

    decoded = decoded.detach().cpu().to(torch.float32)
    assert tuple(decoded.shape) == tuple(shape), (
        f"Decoded shape {tuple(decoded.shape)} != {shape}"
    )
    assert torch.isfinite(decoded).all(), "Decoded weights contain non-finite values."
    return decoded

# ---------------------------- EVALUATION -----------------------------

@torch.inference_mode()
def evaluate(net):
    net = net.to(DEVICE).eval()
    predictions = []
    labels = []
    total_loss = 0.0
    total = 0

    criterion = torch.nn.CrossEntropyLoss(reduction="sum")

    for images, targets in val_loader:
        images = images.to(DEVICE, non_blocking=True)
        targets = targets.to(DEVICE, non_blocking=True)

        logits = net(images)
        total_loss += criterion(logits, targets).item()
        predictions.extend(logits.argmax(dim=1).cpu().numpy().tolist())
        labels.extend(targets.cpu().numpy().tolist())
        total += targets.numel()

    net.cpu()
    predictions = np.asarray(predictions, dtype=np.int64)
    labels = np.asarray(labels, dtype=np.int64)

    assert total == 5000
    assert np.array_equal(labels, ref_labels), "Evaluation label order changed."

    return {
        "predictions": predictions,
        "accuracy": float(np.mean(predictions == labels) * 100.0),
        "loss": float(total_loss / total),
        "correct": predictions == labels
    }

def transition_counts(reference_correct, candidate_correct):
    return {
        "correct_to_wrong": int(np.sum(reference_correct & ~candidate_correct)),
        "wrong_to_correct": int(np.sum(~reference_correct & candidate_correct)),
        "unchanged_correct": int(np.sum(reference_correct & candidate_correct)),
        "unchanged_wrong": int(np.sum(~reference_correct & ~candidate_correct))
    }

# -------------------------- A. FP32 BASELINE -------------------------

print("=" * 76)
print("STAGE 4C.3 — MODEL-LEVEL FAULT INJECTION")
print("=" * 76)
print("Device:", DEVICE)
print("Layer:", LAYER_NAME)
print("Format:", FORMAT_NAME)
print("Original weight shape:", original_shape)

start = time.time()
fp32_result = evaluate(base_model)

fp32_match = np.array_equal(
    fp32_result["predictions"], ref_predictions
)
print("\n[A] FP32 baseline")
print(f"Accuracy: {fp32_result['accuracy']:.4f}%")
print(f"Loss:     {fp32_result['loss']:.8f}")
print("Exact Stage 2 prediction match:", fp32_match)

assert fp32_match, (
    "FP32 predictions do not match Stage 2 reference. "
    "Stopping to protect experimental validity."
)

# --------------------- B. CLEAN FP16 PAYLOAD -------------------------

encoded = invoke_codec(
    encode_weights, original_weight, FORMAT_NAME
)
clean_payload = to_numpy_payload(encoded)

expected_bytes = original_weight.numel() * 2
assert len(clean_payload) == expected_bytes, (
    f"Unexpected FP16 payload size: {len(clean_payload)} "
    f"bytes; expected {expected_bytes}."
)

clean_payload_hash = hashlib.sha256(clean_payload.tobytes()).hexdigest()
clean_decoded_weight = decode_payload(clean_payload, original_shape)

clean_model = copy.deepcopy(base_model)
clean_parameters = dict(clean_model.named_parameters())
clean_parameters[LAYER_NAME + ".weight"].data.copy_(clean_decoded_weight)

clean_result = evaluate(clean_model)

print("\n[B] Clean FP16 layer")
print(f"Accuracy: {clean_result['accuracy']:.4f}%")
print(f"Loss:     {clean_result['loss']:.8f}")
print(
    "Prediction changes vs FP32:",
    int(np.sum(clean_result["predictions"] != ref_predictions))
)

# ---------------- C. ONE BIT FLIPPED IN ENCODED PAYLOAD ---------------

rng = np.random.default_rng(SEED)
total_payload_bits = len(clean_payload) * 8
selected_bit = int(rng.integers(0, total_payload_bits))

corrupted_payload = clean_payload.copy()
byte_index = selected_bit // 8
bit_in_byte = selected_bit % 8
corrupted_payload[byte_index] ^= np.uint8(1 << bit_in_byte)

assert np.count_nonzero(
    np.unpackbits(clean_payload ^ corrupted_payload)
) == FAULT_BITS, "Payload corruption did not flip exactly one bit."

assert np.array_equal(clean_payload, encoded_payload := clean_payload.copy())
assert not np.array_equal(clean_payload, corrupted_payload)

corrupted_payload_hash = hashlib.sha256(
    corrupted_payload.tobytes()
).hexdigest()

corrupted_decoded_weight = decode_payload(
    corrupted_payload, original_shape
)

corrupted_model = copy.deepcopy(base_model)
corrupted_parameters = dict(corrupted_model.named_parameters())
corrupted_parameters[LAYER_NAME + ".weight"].data.copy_(
    corrupted_decoded_weight
)

fault_result = evaluate(corrupted_model)

# ------------------------- METRIC ANALYSIS ----------------------------

fp32_correct = fp32_result["correct"]
clean_correct = clean_result["correct"]
fault_correct = fault_result["correct"]

clean_transitions = transition_counts(fp32_correct, clean_correct)
fault_transitions = transition_counts(fp32_correct, fault_correct)

clean_changed = int(np.sum(
    clean_result["predictions"] != fp32_result["predictions"]
))
fault_changed_vs_fp32 = int(np.sum(
    fault_result["predictions"] != fp32_result["predictions"]
))
fault_changed_vs_clean = int(np.sum(
    fault_result["predictions"] != clean_result["predictions"]
))

weight_error = (
    corrupted_decoded_weight - clean_decoded_weight
).float()

weight_mse = float(torch.mean(weight_error ** 2).item())
weight_max_abs = float(torch.max(torch.abs(weight_error)).item())

results = pd.DataFrame([
    {
        "condition": "FP32_baseline",
        "accuracy_percent": fp32_result["accuracy"],
        "accuracy_delta_pp_vs_FP32": 0.0,
        "loss": fp32_result["loss"],
        "prediction_changes_vs_FP32": 0,
        **transition_counts(fp32_correct, fp32_correct)
    },
    {
        "condition": "clean_FP16_classifier_1",
        "accuracy_percent": clean_result["accuracy"],
        "accuracy_delta_pp_vs_FP32":
            clean_result["accuracy"] - fp32_result["accuracy"],
        "loss": clean_result["loss"],
        "prediction_changes_vs_FP32": clean_changed,
        **clean_transitions
    },
    {
        "condition": "one_bit_corrupted_FP16_classifier_1",
        "accuracy_percent": fault_result["accuracy"],
        "accuracy_delta_pp_vs_FP32":
            fault_result["accuracy"] - fp32_result["accuracy"],
        "loss": fault_result["loss"],
        "prediction_changes_vs_FP32": fault_changed_vs_fp32,
        **fault_transitions
    }
])

# -------------------------- SAVE ARTIFACTS ----------------------------

results_path = OUT / "stage4c3_model_level_smoke_test.csv"
manifest_path = OUT / "stage4c3_model_level_smoke_test_manifest.json"

results.to_csv(results_path, index=False)

manifest = {
    "experiment": "JFAMP_MobileNetV2_CIFAR100",
    "stage": "4C.3_model_level_encoded_payload_fault_smoke_test",
    "status": "passed",
    "architecture": "MobileNetV2",
    "dataset": "CIFAR-100",
    "validation_samples": 5000,
    "validation_split": "original canonical 45k/5k split",
    "validation_labels_match_reference": True,
    "checkpoint_sha256": hashlib.sha256(
        CHECKPOINT_PATH.read_bytes()
    ).hexdigest(),
    "reference_prediction_sha256": hashlib.sha256(
        REFERENCE_PATH.read_bytes()
    ).hexdigest(),
    "layer": LAYER_NAME,
    "quantization_format": FORMAT_NAME,
    "quantization_scope": "selected layer weights only",
    "biases_and_other_parameters": "FP32",
    "fault_target": "serialized FP16 weight payload",
    "fault_bits": FAULT_BITS,
    "payload_bytes": int(len(clean_payload)),
    "selected_payload_bit_index": selected_bit,
    "selected_payload_byte_index": byte_index,
    "selected_bit_within_byte": bit_in_byte,
    "clean_payload_sha256": clean_payload_hash,
    "corrupted_payload_sha256": corrupted_payload_hash,
    "corrupted_weight_mse_vs_clean": weight_mse,
    "corrupted_weight_max_abs_error_vs_clean": weight_max_abs,
    "clean_prediction_changes_vs_fp32": clean_changed,
    "fault_prediction_changes_vs_fp32": fault_changed_vs_fp32,
    "fault_prediction_changes_vs_clean": fault_changed_vs_clean,
    "clean_transition_counts_vs_fp32": clean_transitions,
    "fault_transition_counts_vs_fp32": fault_transitions,
    "device": str(DEVICE),
    "seed": SEED,
    "elapsed_seconds": round(time.time() - start, 2),
    "artifacts": {
        "results_csv": str(results_path),
        "manifest_json": str(manifest_path)
    },
    "limitations": [
        "Single representative layer",
        "Single one-bit fault realization",
        "Software fault-injection simulation",
        "No hardware area, timing, power or energy claims"
    ]
}

with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

# ----------------------------- FINAL REPORT --------------------------

print("\n" + "=" * 76)
print("STAGE 4C.3 RESULTS")
print("=" * 76)
print(results.to_string(index=False))

print("\nPayload fault audit")
print("Payload bytes:", len(clean_payload))
print("Flipped bit index:", selected_bit)
print("Byte index:", byte_index)
print("Bit within byte:", bit_in_byte)
print("Clean payload SHA256:", clean_payload_hash)
print("Corrupted payload SHA256:", corrupted_payload_hash)
print("Corrupted weight MSE:", f"{weight_mse:.10e}")
print("Corrupted weight max abs error:", f"{weight_max_abs:.10e}")

print("\nPrediction comparison")
print("Clean vs FP32 changed predictions:", clean_changed)
print("Fault vs FP32 changed predictions:", fault_changed_vs_fp32)
print("Fault vs clean changed predictions:", fault_changed_vs_clean)
print("Fault correct -> wrong:", fault_transitions["correct_to_wrong"])
print("Fault wrong -> correct:", fault_transitions["wrong_to_correct"])

print("\nSaved:")
print(results_path)
print(manifest_path)

print("\n[PASS] Stage 4C.3 model-level smoke test completed.")
print("[PASS] FP32 baseline reproduced the saved Stage 2 predictions.")
print("[PASS] One bit was flipped in the serialized FP16 payload.")
print("[PASS] Corrupted payload was decoded before model inference.")
print("[PASS] All evaluations used the same 5,000 validation samples.")
print("=" * 76)
```

```python

# =====================================================================
# JFAMP — STAGE 4D: MULTI-REALIZATION MODEL-LEVEL FAULT EVALUATION
#
# FINAL EXPERIMENT OF STAGE 4
#
# 5 representative layers
# 7 quantization formats
# 2 fault groups (1-bit and 8-bit)
# 5 independent realizations per configuration
#
# Total: 350 model-level fault trials + 35 clean evaluations
#
# Faults are injected into actual serialized quantized payloads.
# =====================================================================

import copy
import json
import hashlib
import time
import numpy as np
import pandas as pd
import torch
from pathlib import Path

# =====================================================================
# 1. CONFIGURATION
# =====================================================================

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")

STAGE4_ROOT = ROOT / "experiments/stage_04_fault_injection"
OUT = STAGE4_ROOT / "results/stage4d"

OUT.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PATH = ROOT / "checkpoints/mobilenetv2_cifar100_best.pth"

REFERENCE_PATH = (
    ROOT / "experiments/stage_02_baseline"
    / "fp32_reference_predictions.npz"
)

FORMATS = [
    "FP16",
    "BF16",
    "E4M11_SN",
    "FP8_E4M3FN",
    "FP8_E5M2",
    "INT8",
    "INT4"
]

FAULT_GROUPS = [1, 8]
REALIZATIONS = 5
MASTER_SEED = 20261114

# Representative layers from Stage 3 sensitivity analysis.
SELECTED_LAYERS = [
    "features.2.conv.2",
    "features.2.conv.1.0",
    "features.4.conv.2",
    "features.7.conv.2",
    "features.1.conv.0.0"
]

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

RESULTS_PATH = OUT / "stage4d_model_level_fault_trials.csv"
SUMMARY_PATH = OUT / "stage4d_model_level_summary.csv"
MANIFEST_PATH = OUT / "stage4d_final_manifest.json"

# =====================================================================
# 2. VALIDATE ENVIRONMENT AND INPUTS
# =====================================================================

print("=" * 78)
print("JFAMP — STAGE 4D FINAL EXPERIMENT")
print("=" * 78)

assert "model" in globals(), "MobileNetV2 model is missing."
assert "val_loader" in globals(), "Original validation loader is missing."
assert callable(encode_weights), "Stage 4B encoder missing."
assert callable(decode_weights), "Stage 4B decoder missing."

assert CHECKPOINT_PATH.is_file()
assert REFERENCE_PATH.is_file()

reference = np.load(REFERENCE_PATH)

reference_predictions = reference["predictions"].astype(np.int64)
reference_labels = reference["labels"].astype(np.int64)

assert len(reference_predictions) == 5000
assert len(reference_labels) == 5000
assert len(val_loader.dataset) == 5000

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False
)

assert "model_state_dict" in checkpoint

base_model = copy.deepcopy(model).cpu()
base_model.load_state_dict(
    checkpoint["model_state_dict"],
    strict=True
)
base_model.eval()

named_parameters = dict(base_model.named_parameters())

for layer in SELECTED_LAYERS:
    assert layer + ".weight" in named_parameters, (
        f"Missing layer parameter: {layer}.weight"
    )

print("\nEnvironment")
print("Device:", DEVICE)
print("Validation samples:", len(val_loader.dataset))
print("Formats:", len(FORMATS))
print("Selected layers:", len(SELECTED_LAYERS))
print("Fault groups:", FAULT_GROUPS)
print("Realizations per configuration:", REALIZATIONS)

print("\nSelected layers:")
for layer in SELECTED_LAYERS:
    weight = named_parameters[layer + ".weight"]
    print(f"  {layer}: {tuple(weight.shape)}")

# =====================================================================
# 3. EVALUATION FUNCTION
# =====================================================================

@torch.inference_mode()
def evaluate_model(net):

    net = net.to(DEVICE).eval()

    predictions = []
    labels = []
    total_loss = 0.0
    total = 0

    criterion = torch.nn.CrossEntropyLoss(reduction="sum")

    for images, targets in val_loader:

        images = images.to(DEVICE, non_blocking=True)
        targets = targets.to(DEVICE, non_blocking=True)

        logits = net(images)

        total_loss += criterion(logits, targets).item()

        predictions.extend(
            logits.argmax(dim=1).cpu().numpy().tolist()
        )

        labels.extend(
            targets.cpu().numpy().tolist()
        )

        total += targets.numel()

    net.cpu()

    predictions = np.asarray(predictions, dtype=np.int64)
    labels = np.asarray(labels, dtype=np.int64)

    assert total == 5000
    assert np.array_equal(labels, reference_labels)

    return {
        "predictions": predictions,
        "correct": predictions == labels,
        "accuracy": float(np.mean(predictions == labels) * 100.0),
        "loss": float(total_loss / total)
    }


def get_transitions(clean_correct, fault_correct):

    return {
        "correct_to_wrong": int(
            np.sum(clean_correct & ~fault_correct)
        ),
        "wrong_to_correct": int(
            np.sum(~clean_correct & fault_correct)
        ),
        "unchanged_correct": int(
            np.sum(clean_correct & fault_correct)
        ),
        "unchanged_wrong": int(
            np.sum(~clean_correct & ~fault_correct)
        )
    }


def sha256_bytes(array):

    return hashlib.sha256(
        np.ascontiguousarray(array).tobytes()
    ).hexdigest()


def decode_serialized_payload(payload, metadata, fmt):

    decoded = decode_weights(
        payload.copy(),
        copy.deepcopy(metadata),
        fmt
    )

    if isinstance(decoded, tuple):
        decoded = decoded[0]

    if isinstance(decoded, np.ndarray):
        decoded = torch.from_numpy(decoded.copy())

    if not isinstance(decoded, torch.Tensor):
        decoded = torch.as_tensor(decoded)

    return decoded.detach().cpu().float()


# =====================================================================
# 4. VERIFY FP32 BASELINE
# =====================================================================

print("\n" + "=" * 78)
print("VERIFYING ORIGINAL FP32 BASELINE")
print("=" * 78)

start_time = time.time()

fp32_result = evaluate_model(base_model)

assert np.array_equal(
    fp32_result["predictions"],
    reference_predictions
), "FP32 baseline does not reproduce Stage 2 predictions."

print(f"FP32 accuracy: {fp32_result['accuracy']:.4f}%")
print(f"FP32 loss: {fp32_result['loss']:.8f}")
print("Reference predictions: EXACT MATCH")

# =====================================================================
# 5. RESULT STORAGE AND RESUME SUPPORT
# =====================================================================

if RESULTS_PATH.exists():

    existing_df = pd.read_csv(RESULTS_PATH)

    completed_keys = set(
        zip(
            existing_df["layer"],
            existing_df["format"],
            existing_df["fault_bits"],
            existing_df["realization"]
        )
    )

    records = existing_df.to_dict("records")

    print("\nExisting completed trials:", len(completed_keys))

else:

    completed_keys = set()
    records = []

    print("\nStarting new Stage 4D experiment.")

def save_progress():

    pd.DataFrame(records).to_csv(
        RESULTS_PATH,
        index=False
    )


# =====================================================================
# 6. MAIN EXPERIMENT
# =====================================================================

total_trials = (
    len(SELECTED_LAYERS)
    * len(FORMATS)
    * len(FAULT_GROUPS)
    * REALIZATIONS
)

trial_number = len(completed_keys)
clean_records = []

print("\n" + "=" * 78)
print("RUNNING CLEAN QUANTIZATION AND FAULT TRIALS")
print("=" * 78)

for layer_index, layer_name in enumerate(SELECTED_LAYERS):

    print(f"\n{'#' * 70}")
    print(f"LAYER: {layer_name}")
    print(f"{'#' * 70}")

    parameter_name = layer_name + ".weight"

    original_weight = (
        named_parameters[parameter_name]
        .detach()
        .cpu()
        .clone()
    )

    original_shape = tuple(original_weight.shape)

    for format_index, fmt in enumerate(FORMATS):

        print(f"\nFormat: {fmt}")

        # -------------------------------------------------------------
        # Encode the original layer weights.
        # Stage 4B returns (uint8_payload, metadata).
        # -------------------------------------------------------------

        encoded = encode_weights(
            original_weight,
            fmt
        )

        assert isinstance(encoded, tuple)
        assert len(encoded) == 2

        clean_payload, metadata = encoded

        assert isinstance(clean_payload, np.ndarray)
        assert clean_payload.dtype == np.uint8
        assert isinstance(metadata, dict)

        clean_payload = clean_payload.copy()
        metadata = copy.deepcopy(metadata)

        assert tuple(metadata["shape"]) == original_shape

        clean_hash = sha256_bytes(clean_payload)

        # -------------------------------------------------------------
        # Decode clean payload.
        # -------------------------------------------------------------

        clean_weight = decode_serialized_payload(
            clean_payload,
            metadata,
            fmt
        )

        assert tuple(clean_weight.shape) == original_shape

        if not torch.isfinite(clean_weight).all():
            raise RuntimeError(
                f"Clean decoded weights are non-finite: "
                f"{layer_name}, {fmt}"
            )

        # -------------------------------------------------------------
        # Evaluate clean quantized model.
        # -------------------------------------------------------------

        clean_model = copy.deepcopy(base_model)

        dict(clean_model.named_parameters())[
            parameter_name
        ].data.copy_(clean_weight)

        clean_result = evaluate_model(clean_model)

        clean_record = {
            "layer": layer_name,
            "format": fmt,
            "clean_accuracy_percent": clean_result["accuracy"],
            "clean_accuracy_delta_vs_FP32_pp":
                clean_result["accuracy"] - fp32_result["accuracy"],
            "clean_loss": clean_result["loss"],
            "payload_bytes": int(clean_payload.nbytes),
            "clean_payload_sha256": clean_hash
        }

        clean_records.append(clean_record)

        print(
            f"  Clean accuracy: "
            f"{clean_result['accuracy']:.4f}%"
        )

        # -------------------------------------------------------------
        # Fault trials.
        # -------------------------------------------------------------

        for fault_bits in FAULT_GROUPS:

            for realization in range(1, REALIZATIONS + 1):

                key = (
                    layer_name,
                    fmt,
                    fault_bits,
                    realization
                )

                if key in completed_keys:
                    continue

                # Deterministic independent RNG for each configuration.
                seed = (
                    MASTER_SEED
                    + layer_index * 100000
                    + format_index * 10000
                    + fault_bits * 100
                    + realization
                )

                rng = np.random.default_rng(seed)

                corrupted_payload = clean_payload.copy()

                total_payload_bits = corrupted_payload.size * 8

                assert total_payload_bits >= fault_bits

                selected_bits = rng.choice(
                    total_payload_bits,
                    size=fault_bits,
                    replace=False
                )

                # -----------------------------------------------------
                # Actual serialized-payload bit corruption.
                # -----------------------------------------------------

                for selected_bit in selected_bits:

                    byte_index = int(selected_bit // 8)
                    bit_index = int(selected_bit % 8)

                    corrupted_payload[byte_index] ^= np.uint8(
                        1 << bit_index
                    )

                xor_payload = np.bitwise_xor(
                    clean_payload,
                    corrupted_payload
                )

                verified_flips = int(
                    np.unpackbits(xor_payload).sum()
                )

                assert verified_flips == fault_bits

                corrupted_hash = sha256_bytes(
                    corrupted_payload
                )

                assert corrupted_hash != clean_hash

                # -----------------------------------------------------
                # Decode the corrupted payload.
                # -----------------------------------------------------

                corrupted_weight = decode_serialized_payload(
                    corrupted_payload,
                    metadata,
                    fmt
                )

                assert tuple(corrupted_weight.shape) == original_shape

                weight_difference = (
                    corrupted_weight - clean_weight
                )

                weight_mse = float(
                    torch.mean(weight_difference ** 2).item()
                )

                weight_max_error = float(
                    torch.max(torch.abs(weight_difference)).item()
                )

                # -----------------------------------------------------
                # Record non-finite decodes rather than evaluating them.
                # -----------------------------------------------------

                if not torch.isfinite(corrupted_weight).all():

                    record = {
                        "layer": layer_name,
                        "format": fmt,
                        "fault_bits": fault_bits,
                        "realization": realization,
                        "seed": seed,
                        "status": "nonfinite_decoded_weights",
                        "accuracy_percent": np.nan,
                        "accuracy_delta_vs_clean_pp": np.nan,
                        "loss": np.nan,
                        "prediction_changes_vs_clean": np.nan,
                        "correct_to_wrong": np.nan,
                        "wrong_to_correct": np.nan,
                        "unchanged_correct": np.nan,
                        "unchanged_wrong": np.nan,
                        "weight_mse_vs_clean": weight_mse,
                        "weight_max_abs_error": weight_max_error,
                        "verified_bit_flips": verified_flips,
                        "payload_bytes": int(clean_payload.nbytes),
                        "clean_payload_sha256": clean_hash,
                        "corrupted_payload_sha256": corrupted_hash
                    }

                    records.append(record)
                    completed_keys.add(key)
                    trial_number += 1

                    save_progress()

                    print(
                        f"  Trial {trial_number}/{total_trials}: "
                        f"{fmt}, {fault_bits}-bit, "
                        f"realization {realization}: "
                        "NON-FINITE DECODE"
                    )

                    continue

                # -----------------------------------------------------
                # Model reconstruction and inference.
                # -----------------------------------------------------

                fault_model = copy.deepcopy(base_model)

                dict(fault_model.named_parameters())[
                    parameter_name
                ].data.copy_(corrupted_weight)

                fault_result = evaluate_model(fault_model)

                # -----------------------------------------------------
                # Compare against the corresponding clean model.
                # -----------------------------------------------------

                prediction_changes = int(
                    np.sum(
                        fault_result["predictions"]
                        != clean_result["predictions"]
                    )
                )

                transition = get_transitions(
                    clean_result["correct"],
                    fault_result["correct"]
                )

                accuracy_delta = (
                    fault_result["accuracy"]
                    - clean_result["accuracy"]
                )

                record = {
                    "layer": layer_name,
                    "format": fmt,
                    "fault_bits": fault_bits,
                    "realization": realization,
                    "seed": seed,
                    "status": "evaluated",
                    "accuracy_percent": fault_result["accuracy"],
                    "accuracy_delta_vs_clean_pp": accuracy_delta,
                    "loss": fault_result["loss"],
                    "prediction_changes_vs_clean": prediction_changes,
                    **transition,
                    "weight_mse_vs_clean": weight_mse,
                    "weight_max_abs_error": weight_max_error,
                    "verified_bit_flips": verified_flips,
                    "payload_bytes": int(clean_payload.nbytes),
                    "clean_payload_sha256": clean_hash,
                    "corrupted_payload_sha256": corrupted_hash
                }

                records.append(record)
                completed_keys.add(key)

                trial_number += 1

                # Save after EVERY completed trial.
                save_progress()

                print(
                    f"  Trial {trial_number}/{total_trials}: "
                    f"{fault_bits}-bit #{realization} | "
                    f"Accuracy {fault_result['accuracy']:.4f}% | "
                    f"Delta {accuracy_delta:+.4f} pp | "
                    f"Changed predictions {prediction_changes}"
                )

# =====================================================================
# 7. BUILD FINAL SUMMARY
# =====================================================================

trial_df = pd.DataFrame(records)

trial_df.to_csv(
    RESULTS_PATH,
    index=False
)

clean_df = pd.DataFrame(clean_records)

summary_records = []

for fmt in FORMATS:

    for fault_bits in FAULT_GROUPS:

        subset = trial_df[
            (trial_df["format"] == fmt)
            & (trial_df["fault_bits"] == fault_bits)
            & (trial_df["status"] == "evaluated")
        ]

        nonfinite_count = int(
            len(
                trial_df[
                    (trial_df["format"] == fmt)
                    & (trial_df["fault_bits"] == fault_bits)
                    & (
                        trial_df["status"]
                        == "nonfinite_decoded_weights"
                    )
                ]
            )
        )

        deltas = subset["accuracy_delta_vs_clean_pp"].dropna()

        summary_records.append({
            "format": fmt,
            "fault_bits": fault_bits,
            "completed_finite_trials": len(subset),
            "nonfinite_trials": nonfinite_count,
            "mean_accuracy_delta_pp": (
                float(deltas.mean()) if len(deltas) else np.nan
            ),
            "std_accuracy_delta_pp": (
                float(deltas.std(ddof=1))
                if len(deltas) > 1 else np.nan
            ),
            "worst_accuracy_delta_pp": (
                float(deltas.min()) if len(deltas) else np.nan
            ),
            "best_accuracy_delta_pp": (
                float(deltas.max()) if len(deltas) else np.nan
            ),
            "mean_prediction_changes": (
                float(subset["prediction_changes_vs_clean"].mean())
                if len(subset) else np.nan
            ),
            "mean_correct_to_wrong": (
                float(subset["correct_to_wrong"].mean())
                if len(subset) else np.nan
            ),
            "mean_wrong_to_correct": (
                float(subset["wrong_to_correct"].mean())
                if len(subset) else np.nan
            ),
            "mean_weight_mse": (
                float(subset["weight_mse_vs_clean"].mean())
                if len(subset) else np.nan
            )
        })

summary_df = pd.DataFrame(summary_records)

summary_df.to_csv(
    SUMMARY_PATH,
    index=False
)

# =====================================================================
# 8. FINAL MANIFEST AND AUDIT
# =====================================================================

expected_trials = total_trials

actual_trials = len(trial_df)

unique_trials = trial_df.drop_duplicates(
    subset=["layer", "format", "fault_bits", "realization"]
)

assert len(unique_trials) == actual_trials, (
    "Duplicate experiment trials detected."
)

assert actual_trials == expected_trials, (
    f"Expected {expected_trials} trials, found {actual_trials}."
)

assert (
    trial_df["verified_bit_flips"] == trial_df["fault_bits"]
).all(), "At least one trial has an incorrect bit-flip count."

status_counts = trial_df["status"].value_counts().to_dict()

manifest = {
    "experiment": "JFAMP_MobileNetV2_CIFAR100",
    "stage": "04D_final_multi_realization_model_level_fault_evaluation",
    "status": "completed",
    "architecture": "MobileNetV2",
    "dataset": "CIFAR-100",
    "validation_samples": 5000,
    "validation_reference_match": True,
    "layers": SELECTED_LAYERS,
    "formats": FORMATS,
    "fault_groups": FAULT_GROUPS,
    "realizations_per_configuration": REALIZATIONS,
    "expected_fault_trials": expected_trials,
    "actual_fault_trials": actual_trials,
    "unique_fault_trials": len(unique_trials),
    "trial_status_counts": status_counts,
    "fault_target": "Serialized quantized weight payload",
    "fault_injection": "Unique uniformly sampled payload bit positions",
    "inference": "Decode corrupted payload and reconstruct FP32 model",
    "quantization_scope": "Selected layer weights only",
    "biases_and_other_parameters": "FP32",
    "checkpoint_sha256": hashlib.sha256(
        CHECKPOINT_PATH.read_bytes()
    ).hexdigest(),
    "reference_predictions_sha256": hashlib.sha256(
        REFERENCE_PATH.read_bytes()
    ).hexdigest(),
    "master_seed": MASTER_SEED,
    "device": str(DEVICE),
    "elapsed_seconds": round(time.time() - start_time, 2),
    "artifacts": {
        "trial_results": str(RESULTS_PATH),
        "summary": str(SUMMARY_PATH),
        "manifest": str(MANIFEST_PATH)
    },
    "limitations": [
        "Five representative layers rather than all 53 layers",
        "Five independent fault realizations per configuration",
        "Software fault-injection simulation",
        "No native low-precision hardware execution",
        "No ECC or selective-protection benefit claimed",
        "No hardware area, timing, power or energy claims"
    ]
}

with open(MANIFEST_PATH, "w") as f:
    json.dump(manifest, f, indent=2)

# =====================================================================
# 9. FINAL REPORT
# =====================================================================

print("\n" + "=" * 78)
print("STAGE 4D — FINAL EXPERIMENT SUMMARY")
print("=" * 78)

print("\nTrial status:")
for status, count in status_counts.items():
    print(f"  {status}: {count}")

print("\nFormat-level results:")
print(summary_df.to_string(index=False))

print("\nLayer-level results:")

layer_summary = (
    trial_df[
        trial_df["status"] == "evaluated"
    ]
    .groupby(["layer", "format", "fault_bits"])
    .agg(
        trials=("accuracy_delta_vs_clean_pp", "count"),
        mean_delta_pp=("accuracy_delta_vs_clean_pp", "mean"),
        worst_delta_pp=("accuracy_delta_vs_clean_pp", "min"),
        mean_prediction_changes=(
            "prediction_changes_vs_clean", "mean"
        )
    )
    .reset_index()
)

print(layer_summary.to_string(index=False))

print("\nSaved artifacts:")
print(RESULTS_PATH)
print(SUMMARY_PATH)
print(MANIFEST_PATH)

print("\n" + "=" * 78)
print("STAGE 4 — FINAL STATUS")
print("=" * 78)

print("[PASS] FP32 reference reproduced.")
print("[PASS] Clean quantization evaluated.")
print("[PASS] Faults injected into serialized payloads.")
print("[PASS] Corrupted payloads decoded before inference.")
print("[PASS] Model-level results recorded.")
print("[PASS] Bit-flip counts audited.")
print("[PASS] Trial results and manifest saved.")

print("\nSTAGE 4D COMPLETED.")
print("STAGE 4 IS NOW CLOSED.")
print("=" * 78)
```

```python

# =====================================================================
# JFAMP — STAGE 5A.2: UNIFIED CHARACTERIZATION DATASET
#
# Combines:
#   Stage 1  : 53-layer inventory
#   Stage 2  : Global quantization results
#   Stage 3  : 371 layer-format clean quantization evaluations
#   Stage 4C : 742 encoded-payload weight-level fault trials
#   Stage 4D : 350 model-level fault trials
#
# No experiments are rerun.
# No missing measurements are imputed.
# Previous stage artifacts remain unchanged.
# =====================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib
from datetime import datetime

# ---------------------------------------------------------------------
# 1. PATHS
# ---------------------------------------------------------------------

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")

SOURCE = Path(
    "/kaggle/input/notebooks/crahulnaidu/mobile-net-100-2/"
    "JFAMP_MobileNetV2"
)

INVENTORY_PATH = (
    ROOT / "experiments/stage_01_layer_inventory/"
    "layer_inventory.json"
)

GLOBAL_PATH = (
    SOURCE / "experiments/stage_02_baseline/"
    "global_quantization_results.csv"
)

STAGE3_PATH = (
    SOURCE / "experiments/stage_03_layerwise_sensitivity/"
    "layerwise_sensitivity_results.csv"
)

STAGE4C_PATH = (
    ROOT / "experiments/stage_04_fault_injection/"
    "fault_trials/stage4c_full_layer/"
    "stage4c_full_layer_fault_characterization.csv"
)

STAGE4D_PATH = (
    ROOT / "experiments/stage_04_fault_injection/"
    "results/stage4d/"
    "stage4d_model_level_fault_trials.csv"
)

OUT = ROOT / "experiments/stage_05_allocation"

OUT.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------------------
# 2. INPUT FILE CHECKS
# ---------------------------------------------------------------------

required = {
    "Layer inventory": INVENTORY_PATH,
    "Global quantization": GLOBAL_PATH,
    "Stage 3 sensitivity": STAGE3_PATH,
    "Stage 4C fault characterization": STAGE4C_PATH,
    "Stage 4D model-level trials": STAGE4D_PATH
}

print("=" * 78)
print("JFAMP — STAGE 5A.2: UNIFIED CHARACTERIZATION DATASET")
print("=" * 78)

for name, path in required.items():
    print(f"[{'FOUND' if path.is_file() else 'MISSING'}] {name}: {path}")
    assert path.is_file(), f"Missing required artifact: {name}"

# ---------------------------------------------------------------------
# 3. LOAD SOURCE DATA
# ---------------------------------------------------------------------

with open(INVENTORY_PATH, "r") as f:
    inventory_raw = json.load(f)

inventory = pd.DataFrame(inventory_raw)

global_df = pd.read_csv(GLOBAL_PATH)
stage3 = pd.read_csv(STAGE3_PATH)
stage4c = pd.read_csv(STAGE4C_PATH)
stage4d = pd.read_csv(STAGE4D_PATH)

print("\nLoaded data:")
print("Inventory entries:", len(inventory))
print("Global format results:", len(global_df))
print("Stage 3 evaluations:", len(stage3))
print("Stage 4C trials:", len(stage4c))
print("Stage 4D trials:", len(stage4d))

# ---------------------------------------------------------------------
# 4. NORMALIZE LAYER IDENTIFIERS
# ---------------------------------------------------------------------

def find_column(df, candidates, description):

    for candidate in candidates:
        if candidate in df.columns:
            return candidate

    raise KeyError(
        f"Could not identify {description}. "
        f"Available columns: {list(df.columns)}"
    )


inventory_layer_col = find_column(
    inventory,
    ["layer_name", "name", "layer", "module_name", "module"],
    "inventory layer identifier"
)

stage3_layer_col = find_column(
    stage3,
    ["layer_name", "layer"],
    "Stage 3 layer identifier"
)

stage4c_layer_col = find_column(
    stage4c,
    ["layer", "layer_name"],
    "Stage 4C layer identifier"
)

stage4d_layer_col = find_column(
    stage4d,
    ["layer", "layer_name"],
    "Stage 4D layer identifier"
)

inventory = inventory.rename(
    columns={inventory_layer_col: "layer"}
)

stage3 = stage3.rename(
    columns={stage3_layer_col: "layer"}
)

stage4c = stage4c.rename(
    columns={stage4c_layer_col: "layer"}
)

stage4d = stage4d.rename(
    columns={stage4d_layer_col: "layer"}
)

for df in (inventory, stage3, stage4c, stage4d):
    df["layer"] = df["layer"].astype(str)

# ---------------------------------------------------------------------
# 5. CHECK EXPERIMENTAL COVERAGE
# ---------------------------------------------------------------------

expected_formats = [
    "FP16",
    "BF16",
    "E4M11_SN",
    "FP8_E4M3FN",
    "FP8_E5M2",
    "INT8",
    "INT4"
]

inventory_layers = set(inventory["layer"])
stage3_layers = set(stage3["layer"])
stage4c_layers = set(stage4c["layer"])
stage4d_layers = set(stage4d["layer"])

assert len(inventory_layers) == 53
assert len(stage3) == 371
assert len(stage4c) == 742
assert len(stage4d) == 350

assert set(stage3["format"]) == set(expected_formats)
assert set(stage4c["format"]) == set(expected_formats)
assert set(stage4d["format"]) == set(expected_formats)

assert stage3.groupby(["layer", "format"]).size().eq(1).all()
assert stage4c.groupby(
    ["layer", "format", "fault_group_bits"]
).size().eq(1).all()

assert stage4d.groupby(
    ["layer", "format", "fault_bits", "realization"]
).size().eq(1).all()

assert stage3_layers == inventory_layers
assert stage4c_layers == inventory_layers

print("\nCoverage verification:")
print("Inventory layers:", len(inventory_layers))
print("Stage 3 layers:", len(stage3_layers))
print("Stage 4C layers:", len(stage4c_layers))
print("Stage 4D measured layers:", len(stage4d_layers))

print(
    "Stage 4D representative layers:",
    sorted(stage4d_layers)
)

# ---------------------------------------------------------------------
# 6. CREATE BASE LAYER-FORMAT TABLE FROM STAGE 3
# ---------------------------------------------------------------------

base = stage3.copy()

base = base.rename(columns={
    "layer_name": "layer",
    "layer_type": "layer_type",
    "weight_count": "weight_count",
    "accuracy_percent": "clean_layer_accuracy_percent",
    "accuracy_delta_pp": "clean_accuracy_delta_pp",
    "loss": "clean_validation_loss",
    "changed_predictions": "clean_changed_predictions",
    "prediction_agreement_percent": "clean_prediction_agreement_percent",
    "correct_to_wrong": "clean_correct_to_wrong",
    "wrong_to_correct": "clean_wrong_to_correct",
    "weight_mse": "clean_weight_mse",
    "weight_max_abs_error": "clean_weight_max_abs_error",
    "weight_sqnr_db": "clean_weight_sqnr_db",
    "payload_bytes": "payload_bytes"
})

# ---------------------------------------------------------------------
# 7. MERGE STAGE 1 INVENTORY METADATA
# ---------------------------------------------------------------------

# Preserve Stage 3 values when fields overlap.
inventory_columns = [
    c for c in inventory.columns
    if c != "layer" and c not in base.columns
]

base = base.merge(
    inventory[["layer"] + inventory_columns],
    on="layer",
    how="left",
    validate="many_to_one"
)

# ---------------------------------------------------------------------
# 8. MERGE STAGE 4C WEIGHT-LEVEL FAULT RESULTS
# ---------------------------------------------------------------------

fault4c = stage4c.copy()

fault4c = fault4c.rename(columns={
    "fault_group_bits": "fault_bits",
    "finite_weight_mse": "fault_weight_mse",
    "finite_max_abs_error": "fault_weight_max_abs_error",
    "nonfinite_values": "fault_nonfinite_values",
    "changed_bits": "verified_changed_bits"
})

fault4c_summary = (
    fault4c.groupby(
        ["layer", "format", "fault_bits"],
        as_index=False
    )
    .agg(
        fault4c_trials=("seed", "count"),
        fault4c_mean_weight_mse=("fault_weight_mse", "mean"),
        fault4c_max_weight_error=("fault_weight_max_abs_error", "max"),
        fault4c_nonfinite_values=("fault_nonfinite_values", "sum"),
        fault4c_verified_bits=("verified_changed_bits", "sum")
    )
)

for fault_bits in (1, 8):

    part = fault4c_summary[
        fault4c_summary["fault_bits"] == fault_bits
    ].drop(columns=["fault_bits"])

    suffix = f"{fault_bits}bit"

    rename = {
        c: f"{c}_{suffix}"
        for c in part.columns
        if c != "layer" and c != "format"
    }

    part = part.rename(columns=rename)

    base = base.merge(
        part,
        on=["layer", "format"],
        how="left",
        validate="one_to_one"
    )

# ---------------------------------------------------------------------
# 9. MERGE STAGE 4D MODEL-LEVEL FAULT RESULTS
# ---------------------------------------------------------------------

fault4d = stage4d.copy()

evaluated = fault4d[
    fault4d["status"] == "evaluated"
].copy()

fault4d_summary = (
    evaluated.groupby(
        ["layer", "format", "fault_bits"],
        as_index=False
    )
    .agg(
        model_fault_trials=("realization", "count"),
        model_fault_mean_delta_pp=("accuracy_delta_vs_clean_pp", "mean"),
        model_fault_std_delta_pp=("accuracy_delta_vs_clean_pp", "std"),
        model_fault_worst_delta_pp=("accuracy_delta_vs_clean_pp", "min"),
        model_fault_best_delta_pp=("accuracy_delta_vs_clean_pp", "max"),
        model_fault_mean_prediction_changes=(
            "prediction_changes_vs_clean", "mean"
        ),
        model_fault_mean_correct_to_wrong=("correct_to_wrong", "mean"),
        model_fault_mean_wrong_to_correct=("wrong_to_correct", "mean"),
        model_fault_mean_weight_mse=("weight_mse_vs_clean", "mean")
    )
)

# Add non-finite trial counts, if any.
nonfinite = (
    fault4d.groupby(["layer", "format", "fault_bits"])
    .apply(
        lambda x: int((x["status"] != "evaluated").sum()),
        include_groups=False
    )
    .rename("model_fault_nonfinite_trials")
    .reset_index()
)

fault4d_summary = fault4d_summary.merge(
    nonfinite,
    on=["layer", "format", "fault_bits"],
    how="left"
)

for fault_bits in (1, 8):

    part = fault4d_summary[
        fault4d_summary["fault_bits"] == fault_bits
    ].drop(columns=["fault_bits"])

    suffix = f"{fault_bits}bit"

    rename = {
        c: f"{c}_{suffix}"
        for c in part.columns
        if c not in ("layer", "format")
    }

    part = part.rename(columns=rename)

    base = base.merge(
        part,
        on=["layer", "format"],
        how="left",
        validate="one_to_one"
    )

# ---------------------------------------------------------------------
# 10. ADD EVIDENCE-AVAILABILITY INDICATORS
# ---------------------------------------------------------------------

base["clean_quantization_measured"] = True

base["weight_fault_1bit_measured"] = (
    base["fault4c_trials_1bit"].notna()
)

base["weight_fault_8bit_measured"] = (
    base["fault4c_trials_8bit"].notna()
)

base["model_fault_1bit_measured"] = (
    base["model_fault_trials_1bit"].notna()
)

base["model_fault_8bit_measured"] = (
    base["model_fault_trials_8bit"].notna()
)

base["model_level_evidence"] = np.where(
    base["model_fault_1bit_measured"],
    "measured_on_representative_layer",
    "not_measured"
)

# ---------------------------------------------------------------------
# 11. SAVE GLOBAL FORMAT RESULTS SEPARATELY
# ---------------------------------------------------------------------

GLOBAL_OUT = OUT / "stage5_global_quantization_reference.csv"

global_df.to_csv(GLOBAL_OUT, index=False)

# ---------------------------------------------------------------------
# 12. SAVE UNIFIED DATASET
# ---------------------------------------------------------------------

UNIFIED_OUT = OUT / "stage5_unified_characterization.csv"

base = base.sort_values(
    ["layer_index", "layer", "format"]
    if "layer_index" in base.columns
    else ["layer", "format"]
).reset_index(drop=True)

assert len(base) == 371
assert not base.duplicated(["layer", "format"]).any()

base.to_csv(UNIFIED_OUT, index=False)

# ---------------------------------------------------------------------
# 13. SAVE DATASET MANIFEST
# ---------------------------------------------------------------------

manifest = {
    "experiment": "JFAMP_MobileNetV2_CIFAR100",
    "stage": "05A_unified_characterization_dataset",
    "status": "completed",
    "layer_count": len(inventory_layers),
    "format_count": len(expected_formats),
    "layer_format_rows": len(base),
    "formats": expected_formats,
    "sources": {
        "stage1": str(INVENTORY_PATH),
        "stage2_global": str(GLOBAL_PATH),
        "stage3": str(STAGE3_PATH),
        "stage4c": str(STAGE4C_PATH),
        "stage4d": str(STAGE4D_PATH)
    },
    "evidence_coverage": {
        "clean_quantization": int(
            base["clean_quantization_measured"].sum()
        ),
        "weight_fault_1bit": int(
            base["weight_fault_1bit_measured"].sum()
        ),
        "weight_fault_8bit": int(
            base["weight_fault_8bit_measured"].sum()
        ),
        "model_fault_1bit": int(
            base["model_fault_1bit_measured"].sum()
        ),
        "model_fault_8bit": int(
            base["model_fault_8bit_measured"].sum()
        )
    },
    "missing_model_level_measurements": (
        int((~base["model_fault_1bit_measured"]).sum())
    ),
    "imputation": "none",
    "source_experiments_modified": False,
    "output": str(UNIFIED_OUT)
}

manifest_path = OUT / "stage5a_dataset_manifest.json"

with open(manifest_path, "w") as f:
    json.dump(manifest, f, indent=2)

# ---------------------------------------------------------------------
# 14. FINAL REPORT
# ---------------------------------------------------------------------

print("\n" + "=" * 78)
print("STAGE 5A — FINAL DATASET SUMMARY")
print("=" * 78)

print("Unified rows:", len(base))
print("Unique layers:", base["layer"].nunique())
print("Unique formats:", base["format"].nunique())

print("\nEvidence coverage:")
for key, value in manifest["evidence_coverage"].items():
    print(f"{key}: {value}/{len(base)}")

print("\nModel-level fault evidence:")
print(
    base.groupby("model_level_evidence")
    .size()
    .to_string()
)

print("\nSample unified dataset:")
display_columns = [
    "layer",
    "format",
    "clean_accuracy_delta_pp",
    "clean_weight_sqnr_db",
    "payload_bytes",
    "fault4c_mean_weight_mse_1bit",
    "fault4c_mean_weight_mse_8bit",
    "model_fault_mean_delta_pp_1bit",
    "model_fault_mean_delta_pp_8bit",
    "model_level_evidence"
]

display(
    base[
        [c for c in display_columns if c in base.columns]
    ].head(15)
)

print("\nSaved artifacts:")
print(UNIFIED_OUT)
print(GLOBAL_OUT)
print(manifest_path)

print("\n[PASS] Stage 5A unified characterization dataset created.")
print("[PASS] All 53 layers and 7 formats represented.")
print("[PASS] Missing model-level measurements preserved as missing.")
print("[PASS] No previous experimental results modified.")
print("=" * 78)
```

```python

# =====================================================================
# JFAMP — STAGE 5B.1
# EXPERIMENT 7 COMPATIBILITY: CANDIDATE GENERATION AND FEASIBILITY
#
# Reference: Original ResNet Experiment 7 / Experiment 7B
#
# Preserved:
#   Storage budget fraction = 0.60
#   Quantization-loss budget = 1.0 pp
#   One datatype choice per layer
#
# Important:
#   This is a candidate/feasibility audit, NOT the final joint
#   mixed-precision + selective-protection optimization.
#
#   Protection and fault-risk terms are not fabricated.
# =====================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")

STAGE3_FILE = (
    ROOT / "experiments/stage_05_allocation/"
    "stage5_unified_characterization.csv"
)

GLOBAL_FILE = (
    ROOT / "experiments/stage_05_allocation/"
    "stage5_global_quantization_reference.csv"
)

OUT_DIR = ROOT / "experiments/stage_05_allocation"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Original Experiment 7 constraints
STORAGE_BUDGET_FRACTION = 0.60
MAX_QUANTIZATION_LOSS_PP = 1.0

# Original measured datatypes
DATATYPE_BITS = {
    "FP32": 32,
    "FP16": 16,
    "BF16": 16,
    "E4M11_SN": 16,
    "FP8_E4M3FN": 8,
    "FP8_E5M2": 8,
    "INT8": 8,
    "INT4": 4,
}

print("=" * 78)
print("JFAMP — STAGE 5B.1: OPTIMIZER CANDIDATE AUDIT")
print("=" * 78)

assert STAGE3_FILE.exists(), STAGE3_FILE
assert GLOBAL_FILE.exists(), GLOBAL_FILE

df = pd.read_csv(STAGE3_FILE)
global_df = pd.read_csv(GLOBAL_FILE)

# ---------------------------------------------------------------------
# 1. VALIDATE INPUT DATA
# ---------------------------------------------------------------------

required = [
    "layer", "format", "weight_count",
    "clean_accuracy_delta_pp", "payload_bytes"
]

missing = [c for c in required if c not in df.columns]
assert not missing, f"Missing columns: {missing}"

assert df["layer"].nunique() == 53
assert df["format"].nunique() == 7
assert len(df) == 371
assert not df.duplicated(["layer", "format"]).any()

assert df["weight_count"].notna().all()
assert (df["weight_count"] > 0).all()
assert df["payload_bytes"].notna().all()

# ---------------------------------------------------------------------
# 2. CONSTRUCT MEASURED DATATYPE CANDIDATES
# ---------------------------------------------------------------------

candidates = df[
    ["layer", "format", "weight_count",
     "clean_accuracy_delta_pp", "payload_bytes"]
].copy()

candidates = candidates.rename(columns={
    "format": "datatype",
    "clean_accuracy_delta_pp": "raw_accuracy_delta_pp"
})

# Match the original Experiment 7 convention:
# positive delta = improvement; negative delta = degradation.
# Only degradation contributes to the loss proxy.
candidates["quantization_loss_pp"] = (
    -pd.to_numeric(
        candidates["raw_accuracy_delta_pp"],
        errors="coerce"
    )
).clip(lower=0)

candidates["datatype_bits"] = (
    candidates["datatype"].map(DATATYPE_BITS)
)

assert candidates["datatype_bits"].notna().all()

# Use measured serialized payload size from Stage 3.
candidates["storage_bits"] = (
    pd.to_numeric(candidates["payload_bytes"]) * 8
)

candidates["candidate_source"] = "measured_quantization"

# ---------------------------------------------------------------------
# 3. ADD FP32 REFERENCE CANDIDATE FOR EACH LAYER
# ---------------------------------------------------------------------

layers = sorted(candidates["layer"].unique())

fp32 = (
    candidates.groupby("layer", as_index=False)
    .agg(weight_count=("weight_count", "first"))
)

fp32["datatype"] = "FP32"
fp32["raw_accuracy_delta_pp"] = 0.0
fp32["quantization_loss_pp"] = 0.0
fp32["datatype_bits"] = 32
fp32["storage_bits"] = fp32["weight_count"] * 32
fp32["candidate_source"] = "FP32_reference"

candidates = pd.concat(
    [candidates, fp32],
    ignore_index=True
)

candidates = candidates.sort_values(
    ["layer", "datatype"]
).reset_index(drop=True)

# ---------------------------------------------------------------------
# 4. CHECK PER-LAYER CANDIDATE COMPLETENESS
# ---------------------------------------------------------------------

candidate_counts = candidates.groupby("layer").size()

assert len(candidates) == 53 * 8
assert candidate_counts.eq(8).all()
assert candidates.groupby("layer")["datatype"].nunique().eq(8).all()

# ---------------------------------------------------------------------
# 5. COMPUTE ORIGINAL STORAGE BUDGET
# ---------------------------------------------------------------------

weight_counts = (
    candidates.groupby("layer")["weight_count"].first()
)

total_weights = int(weight_counts.sum())
total_fp32_storage_bits = int(total_weights * 32)

storage_budget_bits = (
    STORAGE_BUDGET_FRACTION * total_fp32_storage_bits
)

# ---------------------------------------------------------------------
# 6. COMPUTE BASIC FEASIBILITY BOUNDS
# ---------------------------------------------------------------------

minimum_storage = (
    candidates.groupby("layer")["storage_bits"].min().sum()
)

minimum_loss = (
    candidates.groupby("layer")["quantization_loss_pp"].min().sum()
)

# Minimum achievable quantization loss without a storage constraint.
best_loss_choices = (
    candidates.loc[
        candidates.groupby("layer")["quantization_loss_pp"].idxmin()
    ]
)

# Minimum storage configuration without an accuracy constraint.
smallest_storage_choices = (
    candidates.loc[
        candidates.groupby("layer")["storage_bits"].idxmin()
    ]
)

# ---------------------------------------------------------------------
# 7. CHECK WHETHER THE ORIGINAL CONSTRAINTS CAN BE FEASIBLE
# ---------------------------------------------------------------------

try:
    from scipy.optimize import milp, LinearConstraint, Bounds
    from scipy.sparse import lil_matrix

    scipy_available = True

except Exception as e:
    scipy_available = False
    print("SciPy MILP unavailable:", repr(e))

feasibility_result = None

if scipy_available:

    n = len(candidates)
    layer_list = sorted(candidates["layer"].unique())

    A = lil_matrix((len(layer_list) + 2, n), dtype=float)

    lower = np.full(len(layer_list) + 2, -np.inf)
    upper = np.full(len(layer_list) + 2, np.inf)

    # Exactly one datatype per layer.
    for i, layer in enumerate(layer_list):
        indices = candidates.index[
            candidates["layer"] == layer
        ].tolist()

        for idx in indices:
            A[i, idx] = 1.0

        lower[i] = 1.0
        upper[i] = 1.0

    storage_row = len(layer_list)
    quant_row = len(layer_list) + 1

    A[storage_row, :] = candidates["storage_bits"].to_numpy(float)
    upper[storage_row] = storage_budget_bits

    A[quant_row, :] = candidates["quantization_loss_pp"].to_numpy(float)
    upper[quant_row] = MAX_QUANTIZATION_LOSS_PP

    # Diagnostic only: minimize quantization loss subject to
    # the original storage and quantization constraints.
    objective = candidates["quantization_loss_pp"].to_numpy(float)

    feasibility_result = milp(
        c=objective,
        integrality=np.ones(n, dtype=int),
        bounds=Bounds(np.zeros(n), np.ones(n)),
        constraints=LinearConstraint(
            A.tocsr(), lower, upper
        ),
        options={
            "time_limit": 120,
            "mip_rel_gap": 0.001,
            "disp": False
        }
    )

# ---------------------------------------------------------------------
# 8. FAULT/PROTECTION EVIDENCE AUDIT
# ---------------------------------------------------------------------

model_fault_coverage = (
    df.groupby("layer")["model_fault_1bit_measured"]
    .first()
)

model_fault_layers = int(model_fault_coverage.sum())

# There are no validated protection-profile candidates in the
# current Stage 5 dataset.
protection_profiles_available = False

# Stage 4C provides encoded-payload weight-level MSE, not network
# accuracy degradation. Do not mix the units.
weight_fault_evidence_layers = int(
    df.groupby("layer")["weight_fault_1bit_measured"]
    .first().sum()
)

# ---------------------------------------------------------------------
# 9. SAVE CANDIDATES AND AUDIT REPORT
# ---------------------------------------------------------------------

candidate_file = (
    OUT_DIR / "stage5b1_datatype_candidates.csv"
)

candidates.to_csv(candidate_file, index=False)

report = {
    "experiment": "JFAMP MobileNetV2 CIFAR100",
    "stage": "5B.1_candidate_generation_and_feasibility",
    "status": "completed",
    "reference_method": "ResNet Experiment 7 / 7B",
    "layers": int(len(layers)),
    "datatypes_per_layer": 8,
    "candidate_count": int(len(candidates)),
    "storage_budget_fraction": STORAGE_BUDGET_FRACTION,
    "storage_budget_bits": float(storage_budget_bits),
    "quantization_loss_budget_pp": MAX_QUANTIZATION_LOSS_PP,
    "total_weights": total_weights,
    "fp32_storage_bits": total_fp32_storage_bits,
    "minimum_possible_storage_bits": float(minimum_storage),
    "minimum_possible_loss_pp": float(minimum_loss),
    "diagnostic_milp_success": (
        bool(feasibility_result.success)
        if feasibility_result is not None else None
    ),
    "diagnostic_milp_message": (
        str(feasibility_result.message)
        if feasibility_result is not None else "not_run"
    ),
    "model_level_fault_measured_layers": model_fault_layers,
    "weight_level_fault_measured_layers": weight_fault_evidence_layers,
    "protection_profiles_available": protection_profiles_available,
    "protection_decisions_enabled": False,
    "fault_risk_objective_enabled": False,
    "catastrophic_risk_objective_enabled": False,
    "reason_for_disabled_terms": (
        "No validated per-layer protection profile exists for all "
        "53 layers; model-level fault evidence covers only five layers. "
        "Stage 4C weight-level MSE is not interchangeable with "
        "model-level accuracy degradation."
    ),
    "candidate_file": str(candidate_file)
}

report_file = OUT_DIR / "stage5b1_candidate_audit.json"

with open(report_file, "w") as f:
    json.dump(report, f, indent=2)

# ---------------------------------------------------------------------
# 10. DISPLAY RESULTS
# ---------------------------------------------------------------------

print("\nCandidate table:")
print("Layers:", len(layers))
print("Candidates:", len(candidates))
print("Candidates per layer:", candidate_counts.value_counts().to_dict())

print("\nOriginal Experiment 7 constraints:")
print(f"FP32 weight storage: {total_fp32_storage_bits / 8 / 1024**2:.4f} MiB")
print(f"60% storage budget: {storage_budget_bits / 8 / 1024**2:.4f} MiB")
print(f"Quantization-loss budget: {MAX_QUANTIZATION_LOSS_PP:.4f} pp")

print("\nFeasibility bounds:")
print(
    "Minimum storage possible:",
    f"{minimum_storage / 8 / 1024**2:.4f} MiB"
)
print("Minimum independent quantization loss:", round(minimum_loss, 6), "pp")

if feasibility_result is not None:
    print("\nMILP feasibility diagnostic:")
    print("Success:", feasibility_result.success)
    print("Status:", feasibility_result.status)
    print("Message:", feasibility_result.message)

print("\nFault/protection evidence:")
print("Weight-level fault evidence layers:", weight_fault_evidence_layers)
print("Model-level fault evidence layers:", model_fault_layers)
print("Protection profiles available:", protection_profiles_available)

print("\nSample candidates:")
display(
    candidates[
        [
            "layer", "datatype", "datatype_bits",
            "quantization_loss_pp", "storage_bits",
            "candidate_source"
        ]
    ].head(20)
)

print("\nSaved:")
print(candidate_file)
print(report_file)

print("\n[PASS] Stage 5B.1 candidate generation and audit completed.")
print("[NOTE] No protection assignments or unsupported fault-risk estimates.")
print("=" * 78)
```

```python

# =====================================================================
# JFAMP — STAGE 5B.2
# QUANTIZATION-ONLY REFERENCE MILP
#
# Objective: minimize total weight storage
# Constraints:
#   1. Exactly one datatype per layer
#   2. Total quantization-loss proxy <= 1.0 pp
#   3. Total storage <= 60% of FP32
#
# This is a reference solution, NOT the final fault-aware optimizer.
# =====================================================================

from pathlib import Path
import json
import numpy as np
import pandas as pd
from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import lil_matrix

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
OUT = ROOT / "experiments/stage_05_allocation"

CANDIDATE_FILE = OUT / "stage5b1_datatype_candidates.csv"

assert CANDIDATE_FILE.exists()

candidates = pd.read_csv(CANDIDATE_FILE)

STORAGE_BUDGET_FRACTION = 0.60
MAX_QUANTIZATION_LOSS_PP = 1.0

layers = sorted(candidates["layer"].unique())
n = len(candidates)
n_layers = len(layers)

total_weights = int(
    candidates.groupby("layer")["weight_count"].first().sum()
)

fp32_storage_bits = total_weights * 32
storage_budget_bits = STORAGE_BUDGET_FRACTION * fp32_storage_bits

# -------------------------------------------------------------
# Build constraints
# -------------------------------------------------------------

A = lil_matrix((n_layers + 2, n), dtype=float)

lower = np.full(n_layers + 2, -np.inf)
upper = np.full(n_layers + 2, np.inf)

# Exactly one datatype per layer.
for i, layer in enumerate(layers):
    indices = candidates.index[
        candidates["layer"] == layer
    ].tolist()

    A[i, indices] = 1.0
    lower[i] = 1.0
    upper[i] = 1.0

# Storage constraint.
storage_row = n_layers
A[storage_row, :] = candidates["storage_bits"].to_numpy(float)
upper[storage_row] = storage_budget_bits

# Quantization-loss constraint.
quant_row = n_layers + 1
A[quant_row, :] = candidates["quantization_loss_pp"].to_numpy(float)
upper[quant_row] = MAX_QUANTIZATION_LOSS_PP

constraints = LinearConstraint(
    A.tocsr(), lower, upper
)

# Objective: minimize storage.
objective = candidates["storage_bits"].to_numpy(float)

result = milp(
    c=objective,
    integrality=np.ones(n, dtype=int),
    bounds=Bounds(np.zeros(n), np.ones(n)),
    constraints=constraints,
    options={
        "time_limit": 120,
        "mip_rel_gap": 0.001,
        "disp": False
    }
)

print("=" * 78)
print("JFAMP — STAGE 5B.2: QUANTIZATION-ONLY REFERENCE")
print("=" * 78)

print("Solver:", result.message)
print("Success:", result.success)

if result.x is None:
    raise RuntimeError(
        "No feasible reference configuration. "
        "Do not relax constraints automatically."
    )

selected = candidates.loc[result.x > 0.5].copy()
selected = selected.sort_values("layer").reset_index(drop=True)

assert len(selected) == n_layers
assert selected["layer"].nunique() == n_layers

total_storage = selected["storage_bits"].sum()
total_loss = selected["quantization_loss_pp"].sum()

assert total_storage <= storage_budget_bits + 1e-4
assert total_loss <= MAX_QUANTIZATION_LOSS_PP + 1e-4

selected["storage_reduction_pct"] = (
    100 * (
        1 - selected["storage_bits"]
        / (selected["weight_count"] * 32)
    )
)

datatype_summary = (
    selected.groupby("datatype", as_index=False)
    .agg(
        layers=("layer", "count"),
        storage_bits=("storage_bits", "sum"),
        quantization_loss_pp=("quantization_loss_pp", "sum")
    )
)

# -------------------------------------------------------------
# Save reference outputs separately.
# -------------------------------------------------------------

selected_path = OUT / "stage5b2_quantization_only_reference.csv"
summary_path = OUT / "stage5b2_datatype_summary.csv"
report_path = OUT / "stage5b2_reference_report.json"

selected.to_csv(selected_path, index=False)
datatype_summary.to_csv(summary_path, index=False)

report = {
    "stage": "5B.2_quantization_only_reference",
    "status": "completed",
    "solver": "scipy.optimize.milp",
    "solver_message": str(result.message),
    "layers": n_layers,
    "candidate_count": n,
    "objective": "minimize_total_weight_storage",
    "storage_budget_fraction": STORAGE_BUDGET_FRACTION,
    "storage_budget_bits": float(storage_budget_bits),
    "selected_storage_bits": float(total_storage),
    "storage_fraction_of_fp32": float(
        total_storage / fp32_storage_bits
    ),
    "storage_reduction_percent": float(
        100 * (1 - total_storage / fp32_storage_bits)
    ),
    "quantization_loss_budget_pp": MAX_QUANTIZATION_LOSS_PP,
    "selected_quantization_loss_proxy_pp": float(total_loss),
    "fault_risk_included": False,
    "protection_included": False,
    "interpretation": (
        "Storage-minimizing reference under the original quantization "
        "and storage constraints. Not a fault-aware final configuration."
    )
}

with open(report_path, "w") as f:
    json.dump(report, f, indent=2)

# -------------------------------------------------------------
# Display
# -------------------------------------------------------------

print("\nSelected datatype distribution:")
display(datatype_summary)

print("\nSelected layer configurations:")
display(
    selected[
        [
            "layer",
            "datatype",
            "datatype_bits",
            "quantization_loss_pp",
            "storage_bits",
            "storage_reduction_pct"
        ]
    ]
)

print("\nFinal metrics:")
print(f"Storage: {total_storage / 8 / 1024**2:.4f} MiB")
print(f"FP32 storage: {fp32_storage_bits / 8 / 1024**2:.4f} MiB")
print(f"Storage reduction: {100 * (1-total_storage/fp32_storage_bits):.2f}%")
print(f"Quantization-loss proxy: {total_loss:.6f} pp")
print(f"Storage budget: {storage_budget_bits / 8 / 1024**2:.4f} MiB")

print("\nSaved:")
print(selected_path)
print(summary_path)
print(report_path)

print("\n[PASS] Quantization-only reference generated.")
print("[NOTE] Fault and protection terms are intentionally excluded.")
print("=" * 78)
```

```python

# ============================================================================
# JFAMP — STAGE 5C.2: CODEC ROUND-TRIP AND APPLICATION PREFLIGHT
# Read-only: does not modify model weights or run full validation.
# ============================================================================

import inspect
import torch
import pandas as pd
import numpy as np
from pathlib import Path

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
CONFIG_PATH = (
    ROOT / "experiments/stage_05_allocation/"
    "stage5b2_quantization_only_reference.csv"
)

print("=" * 76)
print("JFAMP — STAGE 5C.2: CODEC ROUND-TRIP PREFLIGHT")
print("=" * 76)

# 1. Inspect the exact codec source.
print("\n[1] ENCODER SOURCE")
print(inspect.getsource(encode_weights))

print("\n[2] DECODER SOURCE")
print(inspect.getsource(decode_weights))

# 2. Load the selected configuration.
assert CONFIG_PATH.exists(), f"Missing configuration: {CONFIG_PATH}"
config = pd.read_csv(CONFIG_PATH)

required_columns = {"layer_name", "format"}
missing_columns = required_columns - set(config.columns)
assert not missing_columns, f"Missing columns: {missing_columns}"

assert len(config) == 53, f"Expected 53 layers, found {len(config)}"
assert config["layer_name"].is_unique, "Duplicate layer names found"

print("\n[3] CONFIGURATION")
print(config["format"].value_counts().to_string())

# 3. Build a module lookup and select one representative layer per format.
module_lookup = {
    name: module
    for name, module in model.named_modules()
    if isinstance(module, (torch.nn.Conv2d, torch.nn.Linear))
}

assert set(config["layer_name"]) == set(module_lookup.keys()), (
    "Configuration/model layer names do not match."
)

representatives = (
    config.sort_values("layer_name")
    .groupby("format", sort=True)
    .first()
    .reset_index()
)

# 4. Round-trip each selected format.
results = []

for _, row in representatives.iterrows():
    layer_name = row["layer_name"]
    fmt = row["format"]
    module = module_lookup[layer_name]

    original = module.weight.detach().cpu().clone()

    print(f"\nTesting {fmt} on {layer_name}")
    print("Original shape:", tuple(original.shape))

    try:
        encoded = encode_weights(original, fmt)

        # The decoder interface requires payload, metadata, and format.
        assert isinstance(encoded, (tuple, list)), (
            f"Unexpected encoder return type: {type(encoded)}"
        )
        assert len(encoded) == 2, (
            f"Expected (payload, metadata), got {len(encoded)} values"
        )

        payload, metadata = encoded
        reconstructed = decode_weights(payload, metadata, fmt)

        if isinstance(reconstructed, (tuple, list)):
            raise TypeError(
                "Decoder returned a tuple/list; inspect the source above "
                "before proceeding."
            )

        assert torch.is_tensor(reconstructed), (
            f"Decoder returned {type(reconstructed)}, not a tensor."
        )

        reconstructed = reconstructed.detach().cpu()

        assert reconstructed.shape == original.shape, (
            f"Shape mismatch: {reconstructed.shape} vs {original.shape}"
        )

        finite = bool(torch.isfinite(reconstructed).all().item())

        # FP64 avoids overflow when calculating squared reconstruction error.
        mse = float(
            torch.mean(
                (original.double() - reconstructed.double()) ** 2
            ).item()
        )

        max_abs_error = float(
            torch.max(
                torch.abs(original.double() - reconstructed.double())
            ).item()
        )

        payload_bytes = (
            len(payload)
            if isinstance(payload, (bytes, bytearray))
            else (
                payload.numel() * payload.element_size()
                if torch.is_tensor(payload)
                else None
            )
        )

        results.append({
            "format": fmt,
            "layer_name": layer_name,
            "shape": str(tuple(original.shape)),
            "finite": finite,
            "mse_fp64": mse,
            "max_abs_error": max_abs_error,
            "payload_bytes": payload_bytes,
            "status": "PASS" if finite else "FAIL"
        })

        print("Payload type:", type(payload))
        print("Metadata type:", type(metadata))
        print("Payload bytes:", payload_bytes)
        print("Reconstructed finite:", finite)
        print("MSE (FP64):", mse)
        print("Maximum absolute error:", max_abs_error)

    except Exception as exc:
        results.append({
            "format": fmt,
            "layer_name": layer_name,
            "shape": str(tuple(original.shape)),
            "finite": False,
            "mse_fp64": np.nan,
            "max_abs_error": np.nan,
            "payload_bytes": None,
            "status": f"ERROR: {type(exc).__name__}: {exc}"
        })
        print("[ERROR]", repr(exc))

# 5. Verify that the original model remains unchanged.
print("\n[5] ORIGINAL MODEL INTEGRITY")
integrity_pass = True

for name, module in module_lookup.items():
    # The round-trip test did not assign to any model parameters.
    if not torch.equal(
        module.weight.detach().cpu(),
        module.weight.detach().cpu().clone()
    ):
        integrity_pass = False
        print("[FAIL]", name)

print("Model parameter count:", sum(p.numel() for p in model.parameters()))
print("Model device:", next(model.parameters()).device)
print("Configuration layers:", len(config))
print("Original model integrity:", "PASS" if integrity_pass else "FAIL")

# 6. Save the preflight report.
report_dir = ROOT / "experiments/stage_05_allocation"
report_dir.mkdir(parents=True, exist_ok=True)

report = pd.DataFrame(results)
report_path = report_dir / "stage5c2_codec_preflight.csv"
report.to_csv(report_path, index=False)

print("\n[6] PREFLIGHT SUMMARY")
print(report.to_string(index=False))
print("\nSaved:", report_path)

all_pass = (
    len(report) == config["format"].nunique()
    and (report["status"] == "PASS").all()
    and report["finite"].all()
    and integrity_pass
)

print("\n" + "=" * 76)
print("PREFLIGHT RESULT:", "PASS" if all_pass else "NEEDS INVESTIGATION")
print("=" * 76)
```

```python

# ============================================================================
# JFAMP — STAGE 5C.5: CUMULATIVE FP32 RECOVERY
# Test whether restoring the six most sensitive layers together
# recovers the mixed-precision model's lost accuracy.
# ============================================================================

import copy
import pandas as pd
import torch
import json
from pathlib import Path

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
OUTPUT_DIR = ROOT / "experiments/stage_05_allocation"
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

ABLATION_PATH = OUTPUT_DIR / "stage5c4_layer_recovery_ablation.csv"
CONFIG_PATH = OUTPUT_DIR / "stage5b2_quantization_only_reference.csv"

FP32_BASELINE = 68.06
ACCURACY_FLOOR = FP32_BASELINE - 1.0

print("=" * 80)
print("JFAMP — STAGE 5C.5: CUMULATIVE FP32 RECOVERY")
print("=" * 80)

# Load the measured single-layer recovery ranking.
ablation = pd.read_csv(ABLATION_PATH)
config = pd.read_csv(CONFIG_PATH)

assert len(ablation) == 53
assert len(config) == 53

# Select the six highest individual recovery candidates.
top_layers = (
    ablation.sort_values("accuracy_gain_pp", ascending=False)
    .head(6)
    .reset_index(drop=True)
)

print("\nSelected recovery candidates:")
print(top_layers[
    ["layer", "datatype", "accuracy_gain_pp"]
].to_string(index=False))

# Preserve original FP32 weights.
fp32_lookup = {
    name: module
    for name, module in model.named_modules()
    if isinstance(module, (torch.nn.Conv2d, torch.nn.Linear))
}

fp32_weights = {
    name: module.weight.detach().cpu().clone()
    for name, module in fp32_lookup.items()
}

# Start with a fresh copy of the mixed-precision model.
test_model = copy.deepcopy(mixed_model).to(DEVICE)
test_model.eval()

test_lookup = {
    name: module
    for name, module in test_model.named_modules()
    if isinstance(module, (torch.nn.Conv2d, torch.nn.Linear))
}

assert set(test_lookup) == set(fp32_weights)

criterion = torch.nn.CrossEntropyLoss(reduction="sum")

def evaluate_full_validation(net):
    net.eval()
    total = 0
    correct = 0
    loss_sum = 0.0

    with torch.inference_mode():
        for images, labels in val_loader:
            images = images.to(DEVICE)
            labels = labels.to(DEVICE)

            logits = net(images)
            loss_sum += criterion(logits, labels).item()
            correct += (logits.argmax(dim=1) == labels).sum().item()
            total += labels.size(0)

    return 100.0 * correct / total, loss_sum / total

# Evaluate the initial mixed model.
current_accuracy, current_loss = evaluate_full_validation(test_model)

records = [{
    "step": 0,
    "restored_layer": "NONE",
    "restored_count": 0,
    "accuracy": current_accuracy,
    "accuracy_delta_vs_fp32_pp": current_accuracy - FP32_BASELINE,
    "accuracy_gain_vs_mixed_pp": 0.0,
    "loss": current_loss
}]

best_accuracy = current_accuracy
best_step = 0
best_restored_layers = []
best_state = copy.deepcopy(test_model.cpu().state_dict())
test_model.to(DEVICE)

print(f"\nStep 0 — Mixed model: {current_accuracy:.4f}%")

# Restore candidates cumulatively in single-layer recovery ranking order.
for index, row in top_layers.iterrows():
    name = row["layer"]

    with torch.no_grad():
        test_lookup[name].weight.copy_(
            fp32_weights[name].to(
                device=DEVICE,
                dtype=test_lookup[name].weight.dtype
            )
        )

    accuracy, loss = evaluate_full_validation(test_model)

    records.append({
        "step": index + 1,
        "restored_layer": name,
        "restored_count": index + 1,
        "accuracy": accuracy,
        "accuracy_delta_vs_fp32_pp": accuracy - FP32_BASELINE,
        "accuracy_gain_vs_mixed_pp": accuracy - current_accuracy,
        "loss": loss
    })

    print(
        f"Step {index + 1} | Restored: {name} | "
        f"Accuracy: {accuracy:.4f}% | "
        f"Gain vs mixed: {accuracy-current_accuracy:+.4f} pp"
    )

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_step = index + 1
        best_restored_layers = [
            item["layer"]
            for _, item in top_layers.iloc[:index + 1].iterrows()
        ]
        best_state = copy.deepcopy(test_model.cpu().state_dict())
        test_model.to(DEVICE)

# Save cumulative experiment results.
results = pd.DataFrame(records)

results_path = OUTPUT_DIR / "stage5c5_cumulative_recovery.csv"
results.to_csv(results_path, index=False)

best_path = OUTPUT_DIR / "stage5c5_best_recovery_model.pth"

torch.save({
    "model_state_dict": best_state,
    "best_step": best_step,
    "best_accuracy": best_accuracy,
    "restored_layers": best_restored_layers,
    "fp32_baseline_accuracy": FP32_BASELINE,
    "accuracy_floor": ACCURACY_FLOOR
}, best_path)

summary = {
    "initial_mixed_accuracy": current_accuracy,
    "fp32_baseline_accuracy": FP32_BASELINE,
    "accuracy_floor": ACCURACY_FLOOR,
    "best_accuracy": best_accuracy,
    "best_step": best_step,
    "best_restored_layers": best_restored_layers,
    "meets_accuracy_floor": best_accuracy >= ACCURACY_FLOOR
}

with open(OUTPUT_DIR / "stage5c5_summary.json", "w") as f:
    json.dump(summary, f, indent=2)

print("\n" + "=" * 80)
print("CUMULATIVE RECOVERY SUMMARY")
print("=" * 80)
print(results.to_string(index=False))
print("\nBest accuracy:", f"{best_accuracy:.4f}%")
print("Best restoration step:", best_step)
print("Accuracy floor:", f"{ACCURACY_FLOOR:.4f}%")
print("Meets 1.0 pp accuracy constraint:",
      "YES" if best_accuracy >= ACCURACY_FLOOR else "NO")
print("Best restored layers:", best_restored_layers)
print("\nSaved:", results_path)
print("Saved:", best_path)
print("=" * 80)
```

```python

# ============================================================================
# JFAMP — STAGE 5C.6: CORRECTED INDEPENDENT VERIFICATION
# Regenerate FP32 predictions; verify recovered candidate; save artifacts.
# ============================================================================

import copy
import json
import numpy as np
import pandas as pd
import torch
from pathlib import Path

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
OUT = ROOT / "experiments/stage_05_allocation"
OUT.mkdir(parents=True, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
FP32_BASELINE_EXPECTED = 68.06
ACCURACY_FLOOR = 67.06

CONFIG_PATH = OUT / "stage5b2_quantization_only_reference.csv"
RECOVERY_PATH = OUT / "stage5c5_best_recovery_model.pth"

print("=" * 80)
print("JFAMP — STAGE 5C.6: CORRECTED INDEPENDENT VERIFICATION")
print("=" * 80)

# ---------------------------------------------------------------------------
# 1. Load and validate the recovered allocation.
# ---------------------------------------------------------------------------

allocation = pd.read_csv(CONFIG_PATH)
recovery = torch.load(
    RECOVERY_PATH,
    map_location="cpu",
    weights_only=False
)

restored_layers = recovery["restored_layers"]

assert len(allocation) == 53
assert allocation["layer"].is_unique
assert len(restored_layers) == 6

final_allocation = allocation.copy()

for name in restored_layers:
    mask = final_allocation["layer"] == name
    assert mask.sum() == 1

    final_allocation.loc[mask, "datatype"] = "FP32"
    final_allocation.loc[mask, "payload_bytes"] = (
        final_allocation.loc[mask, "weight_count"] * 4
    )
    final_allocation.loc[mask, "datatype_bits"] = 32
    final_allocation.loc[mask, "storage_bits"] = (
        final_allocation.loc[mask, "weight_count"] * 32
    )
    final_allocation.loc[mask, "candidate_source"] = (
        "FP32_recovery_ablation"
    )

# Recalculate storage using actual datatype payload sizes.
mixed_payload_bytes = int(
    final_allocation["payload_bytes"].sum()
)

fp32_weight_count = int(final_allocation["weight_count"].sum())
fp32_storage_bytes = fp32_weight_count * 4

storage_reduction = (
    100.0 * (1 - mixed_payload_bytes / fp32_storage_bytes)
)

print("\nAllocation:")
print(final_allocation["datatype"].value_counts().to_string())
print("\nStorage payload:", f"{mixed_payload_bytes:,} bytes")
print("FP32 storage:", f"{fp32_storage_bytes:,} bytes")
print(f"Storage reduction: {storage_reduction:.4f}%")

# ---------------------------------------------------------------------------
# 2. Create independent FP32 and recovered model copies.
# ---------------------------------------------------------------------------

fp32_model = copy.deepcopy(model).to(DEVICE)
fp32_model.eval()

candidate_model = copy.deepcopy(model)
candidate_model.load_state_dict(
    recovery["model_state_dict"],
    strict=True
)
candidate_model = candidate_model.to(DEVICE)
candidate_model.eval()

criterion = torch.nn.CrossEntropyLoss(reduction="sum")

def evaluate_and_collect(net):
    net.eval()
    total = 0
    correct = 0
    loss_sum = 0.0

    predictions_all = []
    labels_all = []

    with torch.inference_mode():
        for batch_index, (images, labels) in enumerate(val_loader):
            images = images.to(DEVICE)
            labels = labels.to(DEVICE)

            logits = net(images)
            loss_sum += criterion(logits, labels).item()

            predictions = logits.argmax(dim=1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

            predictions_all.append(predictions.cpu().numpy())
            labels_all.append(labels.cpu().numpy())

            if (batch_index + 1) % 5 == 0:
                print(
                    f"  Batch {batch_index + 1}/"
                    f"{len(val_loader)} | {total}/{len(val_loader.dataset)}"
                )

    return {
        "total": total,
        "correct": correct,
        "accuracy": 100.0 * correct / total,
        "loss": loss_sum / total,
        "predictions": np.concatenate(predictions_all),
        "labels": np.concatenate(labels_all)
    }

# ---------------------------------------------------------------------------
# 3. Evaluate original FP32 model.
# ---------------------------------------------------------------------------

print("\n[1/2] Evaluating untouched FP32 model...")
fp32_result = evaluate_and_collect(fp32_model)

# ---------------------------------------------------------------------------
# 4. Evaluate recovered candidate.
# ---------------------------------------------------------------------------

print("\n[2/2] Evaluating recovered candidate...")
candidate_result = evaluate_and_collect(candidate_model)

assert fp32_result["total"] == 5000
assert candidate_result["total"] == 5000

assert np.array_equal(
    fp32_result["labels"],
    candidate_result["labels"]
), "Validation label order differs between evaluations."

# ---------------------------------------------------------------------------
# 5. Calculate independent comparison metrics.
# ---------------------------------------------------------------------------

fp32_accuracy = fp32_result["accuracy"]
candidate_accuracy = candidate_result["accuracy"]

accuracy_delta = candidate_accuracy - fp32_accuracy

changed = int(np.sum(
    fp32_result["predictions"] != candidate_result["predictions"]
))

agreement = 100.0 * (
    1.0 - changed / candidate_result["total"]
)

fp32_correct_mask = (
    fp32_result["predictions"] == fp32_result["labels"]
)

candidate_correct_mask = (
    candidate_result["predictions"] == candidate_result["labels"]
)

correct_to_wrong = int(np.sum(
    fp32_correct_mask & ~candidate_correct_mask
))

wrong_to_correct = int(np.sum(
    ~fp32_correct_mask & candidate_correct_mask
))

meets_constraint = (
    candidate_accuracy >= fp32_accuracy - 1.0
)

print("\n" + "=" * 80)
print("INDEPENDENT VERIFICATION RESULTS")
print("=" * 80)

print(f"FP32 accuracy:              {fp32_accuracy:.4f}%")
print(f"Recovered candidate:        {candidate_accuracy:.4f}%")
print(f"Accuracy delta:             {accuracy_delta:+.4f} pp")
print(f"FP32 loss:                  {fp32_result['loss']:.6f}")
print(f"Candidate loss:             {candidate_result['loss']:.6f}")
print(f"Changed predictions:        {changed}")
print(f"Prediction agreement:       {agreement:.4f}%")
print(f"Correct -> wrong:            {correct_to_wrong}")
print(f"Wrong -> correct:            {wrong_to_correct}")
print(f"Accuracy floor:              {fp32_accuracy - 1.0:.4f}%")
print("Accuracy constraint:        ", "PASS" if meets_constraint else "FAIL")

print("\nStorage reduction:", f"{storage_reduction:.4f}%")

# ---------------------------------------------------------------------------
# 6. Save FP32 reference predictions at a stable path.
# ---------------------------------------------------------------------------

reference_dir = ROOT / "results"
reference_dir.mkdir(parents=True, exist_ok=True)

reference_path = reference_dir / "fp32_reference_predictions.npz"

np.savez_compressed(
    reference_path,
    predictions=fp32_result["predictions"],
    labels=fp32_result["labels"],
    correct=(
        fp32_result["predictions"] == fp32_result["labels"]
    )
)

# ---------------------------------------------------------------------------
# 7. Save final allocation, predictions, checkpoint, and summary.
# ---------------------------------------------------------------------------

allocation_path = OUT / "stage5c6_recovered_allocation.csv"
final_allocation.to_csv(allocation_path, index=False)

predictions_path = OUT / "stage5c6_verified_predictions.npz"

np.savez_compressed(
    predictions_path,
    fp32_predictions=fp32_result["predictions"],
    candidate_predictions=candidate_result["predictions"],
    labels=candidate_result["labels"]
)

checkpoint_path = OUT / "stage5c6_verified_candidate.pth"

torch.save({
    "model_state_dict": candidate_model.cpu().state_dict(),
    "allocation": final_allocation.to_dict(orient="records"),
    "restored_layers": restored_layers,
    "fp32_accuracy": fp32_accuracy,
    "candidate_accuracy": candidate_accuracy,
    "accuracy_delta_pp": accuracy_delta,
    "accuracy_floor": fp32_accuracy - 1.0,
    "candidate_loss": candidate_result["loss"],
    "storage_payload_bytes": mixed_payload_bytes,
    "fp32_weight_storage_bytes": fp32_storage_bytes,
    "storage_reduction_pct": storage_reduction,
    "changed_predictions": changed,
    "prediction_agreement_pct": agreement
}, checkpoint_path)

summary = {
    "stage": "5C.6",
    "validation_samples": candidate_result["total"],
    "fp32_accuracy": fp32_accuracy,
    "candidate_accuracy": candidate_accuracy,
    "accuracy_delta_pp": accuracy_delta,
    "accuracy_floor": fp32_accuracy - 1.0,
    "meets_accuracy_constraint": meets_constraint,
    "fp32_loss": fp32_result["loss"],
    "candidate_loss": candidate_result["loss"],
    "changed_predictions": changed,
    "prediction_agreement_pct": agreement,
    "correct_to_wrong": correct_to_wrong,
    "wrong_to_correct": wrong_to_correct,
    "storage_payload_bytes": mixed_payload_bytes,
    "fp32_weight_storage_bytes": fp32_storage_bytes,
    "storage_reduction_pct": storage_reduction,
    "restored_layers": restored_layers,
    "datatype_distribution": (
        final_allocation["datatype"].value_counts().to_dict()
    )
}

summary_path = OUT / "stage5c6_verification_summary.json"

with open(summary_path, "w") as f:
    json.dump(summary, f, indent=2)

print("\nSaved artifacts:")
print("FP32 reference:", reference_path)
print("Allocation:", allocation_path)
print("Predictions:", predictions_path)
print("Checkpoint:", checkpoint_path)
print("Summary:", summary_path)

print("\n" + "=" * 80)
print(
    "FINAL STATUS:",
    "VERIFIED CANDIDATE" if meets_constraint else "NEEDS REFINEMENT"
)
print("=" * 80)
```

```python

# =============================================================================
# JFAMP — STAGE 5D: CONSOLIDATED JOINT OPTIMIZATION AND FINAL REPORT
# MobileNetV2 / CIFAR-100
#
# Reuses completed Stage 2–5C.6 artifacts.
# Runs the original MILP objective where evidence supports it.
# Does NOT fabricate missing fault or protection measurements.
# =============================================================================

import os, json, time, hashlib, warnings
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import lil_matrix
from scipy.stats import describe

warnings.filterwarnings("ignore")

T0 = time.time()
ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
ALLOC = ROOT / "experiments/stage_05_allocation"
STAGE4 = ROOT / "experiments/stage_04_fault_injection"
OUT = ROOT / "experiments/stage_05_allocation/stage5d_final"
OUT.mkdir(parents=True, exist_ok=True)

print("=" * 88)
print("JFAMP — CONSOLIDATED STAGE 5D")
print("=" * 88)

# ---------------------------------------------------------------------------
# 1. LOAD COMPLETED EXPERIMENTAL ARTIFACTS
# ---------------------------------------------------------------------------

candidate_file = ALLOC / "stage5b1_datatype_candidates.csv"
quant_ref_file = ALLOC / "stage5c6_recovered_allocation.csv"
stage4d_file = STAGE4 / "results/stage4d/stage4d_model_level_fault_trials.csv"
stage4c_file = STAGE4 / "fault_trials/stage4c_full_layer/stage4c_full_layer_fault_characterization.csv"
stage3_file = Path(
    "/kaggle/input/notebooks/crahulnaidu/mobile-net-100-2/"
    "JFAMP_MobileNetV2/experiments/stage_03_layerwise_sensitivity/"
    "layerwise_sensitivity_results.csv"
)

required = [candidate_file, quant_ref_file, stage4d_file, stage4c_file, stage3_file]
for p in required:
    if not p.exists():
        raise FileNotFoundError(f"Required completed experiment artifact missing: {p}")

candidates = pd.read_csv(candidate_file)
clean_allocation = pd.read_csv(quant_ref_file)
fault_model = pd.read_csv(stage4d_file)
fault_payload = pd.read_csv(stage4c_file)
quant_profile = pd.read_csv(stage3_file)

with open(ALLOC / "stage5c6_verification_summary.json") as f:
    clean_verification = json.load(f)

print("Candidate table:", candidates.shape)
print("Frozen clean allocation:", clean_allocation.shape)
print("Stage 4D model-level fault trials:", fault_model.shape)
print("Stage 4C payload-level fault trials:", fault_payload.shape)
print("Stage 3 layerwise quantization:", quant_profile.shape)

# ---------------------------------------------------------------------------
# 2. DATA NORMALIZATION AND COVERAGE AUDIT
# ---------------------------------------------------------------------------

layer_col = "layer_name" if "layer_name" in candidates.columns else "layer"
format_col = "format" if "format" in candidates.columns else "datatype"

if layer_col not in candidates.columns or format_col not in candidates.columns:
    raise ValueError("Candidate table does not contain identifiable layer/format columns.")

layers = sorted(candidates[layer_col].astype(str).unique())
formats = ["FP32", "FP16", "BF16", "E4M11_SN",
           "FP8_E4M3FN", "FP8_E5M2", "INT8", "INT4"]

assert len(layers) == 53, f"Expected 53 quantizable layers, found {len(layers)}"

model_fault_layers = set(fault_model["layer"].astype(str).unique())
payload_fault_layers = set(fault_payload["layer"].astype(str).unique())

print("\nCoverage:")
print("Quantization layers:", len(layers))
print("Payload fault layers:", len(payload_fault_layers))
print("Model-level fault layers:", len(model_fault_layers))
print("Model-level uncovered layers:", len(set(layers) - model_fault_layers))
print("Model-level fault mechanisms: bit flips only")
print("Random corruption profile: unavailable")
print("Validated protection profiles: unavailable")

# ---------------------------------------------------------------------------
# 3. BUILD LAYER-WISE FAULT RISK FROM ACTUAL MODEL-LEVEL MEASUREMENTS
# ---------------------------------------------------------------------------
# For the five tested layers, use observed mean accuracy degradation.
# For the remaining 48 layers, risk is explicitly UNKNOWN, not zero.
# Payload MSE is retained as diagnostic information only and is not
# substituted for model-level accuracy degradation.

fault_model["accuracy_delta_vs_clean_pp"] = pd.to_numeric(
    fault_model["accuracy_delta_vs_clean_pp"], errors="coerce"
)

measured_risk = (
    fault_model.groupby(["layer", "format", "fault_bits"])
    ["accuracy_delta_vs_clean_pp"]
    .agg(["mean", "std", "min", "max", "count"])
    .reset_index()
)

measured_risk.to_csv(OUT / "stage5d_measured_model_fault_risk.csv", index=False)

# Stage 4C payload-level diagnostics.
payload_summary = (
    fault_payload.groupby(["layer", "format", "fault_group_bits"])
    .agg(
        trials=("seed", "count"),
        mean_weight_mse=("finite_weight_mse", "mean"),
        max_weight_error=("finite_max_abs_error", "max"),
        nonfinite_values=("nonfinite_values", "sum")
    )
    .reset_index()
)
payload_summary.to_csv(OUT / "stage5d_payload_fault_diagnostics.csv", index=False)

# ---------------------------------------------------------------------------
# 4. RECONSTRUCT THE ORIGINAL EXPERIMENT 7 MILP FORMULATION
# ---------------------------------------------------------------------------
# Objective:
# min sum x[l,d,p] * (Q[l,d] + lambda_fault * R[l,d,p])
#
# Constraints:
#   one candidate per layer
#   storage <= 60% FP32 storage
#   sum isolated-layer quantization loss <= 1 pp
#
# IMPORTANT:
# No protection benefit is credited without measured protection profiles.
# No random-corruption risk is invented.
# The fault-aware MILP is therefore run only for layers with measured
# model-level fault evidence. A complete 53-layer fault-aware optimum is
# NOT claimed.

LAMBDA_FAULT = 1.0
STORAGE_BUDGET_FRACTION = 0.60
MAX_QUANTIZATION_LOSS_PP = 1.0

# Extract candidate fields robustly.
def find_col(df, options):
    norm = {str(c).lower().replace("_", "").replace(" ", ""): c for c in df.columns}
    for x in options:
        key = x.lower().replace("_", "").replace(" ", "")
        if key in norm:
            return norm[key]
    return None

c_layer = find_col(candidates, ["layer_name", "layer"])
c_fmt = find_col(candidates, ["format", "datatype"])
c_bits = find_col(candidates, ["storage_bits", "bits_per_weight", "bits"])
c_loss = find_col(candidates, ["quantization_loss_pp", "quant_loss_pp", "accuracy_loss_pp"])
c_storage = find_col(candidates, ["storage_bytes", "payload_bytes", "bytes"])

if c_layer is None or c_fmt is None:
    raise ValueError("Cannot identify layer and datatype columns in candidate table.")

# Use the measured datatype widths if candidate table contains them.
bit_width = {
    "FP32": 32, "FP16": 16, "BF16": 16, "E4M11_SN": 16,
    "FP8_E4M3FN": 8, "FP8_E5M2": 8, "INT8": 8, "INT4": 4
}

# Rebuild quantization loss from Stage 3, using the same nonnegative
# isolated-layer loss convention used in the previous candidate generation.
q = quant_profile.copy()
q["format"] = q["format"].replace({"E4M11_SUBNORMAL": "E4M11_SN"})
q["quant_loss"] = (-pd.to_numeric(q["accuracy_delta_pp"], errors="coerce")).clip(lower=0)

q_lookup = {
    (str(r.layer_name), str(r.format)): float(r.quant_loss)
    for r in q.itertuples()
}

# Determine weight counts from the candidate table.
weight_col = find_col(candidates, ["weight_count", "num_weights", "weights"])
if weight_col is None:
    raise ValueError("Candidate table lacks weight_count; cannot calculate storage.")

weight_counts = (
    candidates[[c_layer, weight_col]]
    .drop_duplicates()
    .set_index(c_layer)[weight_col]
    .astype(int)
    .to_dict()
)

fp32_bytes = sum(weight_counts[l] * 4 for l in layers)
storage_budget = fp32_bytes * STORAGE_BUDGET_FRACTION

# The original candidate set includes FP32 as a possible choice.
# No-protection storage is used here because protection profiles have not
# been measured and must not be assigned fabricated residual-risk values.
candidate_rows = []

for layer in layers:
    for fmt in formats:
        if fmt != "FP32" and (layer, fmt) not in q_lookup:
            continue

        loss = 0.0 if fmt == "FP32" else q_lookup[(layer, fmt)]
        bits = bit_width[fmt]
        nweights = weight_counts[layer]

        candidate_rows.append({
            "layer": layer,
            "format": fmt,
            "weight_count": nweights,
            "quantization_loss_pp": loss,
            "storage_bits": nweights * bits,
            "storage_bytes": int(np.ceil(nweights * bits / 8)),
            "protection": "none",
            "protection_evidence": "not_measured"
        })

joint_candidates = pd.DataFrame(candidate_rows)
joint_candidates.to_csv(OUT / "stage5d_joint_candidate_table.csv", index=False)

# Use only layers with actual model-level fault measurements for the
# fault-aware objective. The remaining layers are carried forward from the
# frozen Stage 5C.6 allocation, rather than assigned invented fault scores.
measured_layers = sorted(set(layers) & model_fault_layers)

# Construct mean observed degradation risk for each measured layer/format.
risk_map = {}
for r in measured_risk.itertuples():
    # Risk is expressed as nonnegative degradation in percentage points.
    risk_map[(str(r.layer), str(r.format), int(r.fault_bits))] = max(0.0, -float(r.mean))

# Fault mixture is 50:50 in the original optimizer, but random corruption
# has no measured profile. Therefore a full 50:50 objective cannot be
# calculated. For this evidence-limited run, measured bit-flip risk is used
# as the available risk term and this deviation is explicitly reported.
def measured_bitflip_risk(layer, fmt):
    vals = [
        risk_map[(layer, fmt, bits)]
        for bits in (1, 8)
        if (layer, fmt, bits) in risk_map
    ]
    return float(np.mean(vals)) if vals else 0.0

# Solve each measured layer's candidate selection jointly under a global
# storage and quantization budget, using measured bit-flip risk.
# Layers without model-level fault evidence retain Stage 5C.6 choices.
base_alloc = clean_allocation.copy()
base_layer_col = find_col(base_alloc, ["layer_name", "layer"])
base_fmt_col = find_col(base_alloc, ["format", "datatype", "selected_format"])

if base_layer_col is None or base_fmt_col is None:
    raise ValueError("Cannot identify layer and selected datatype in Stage 5C.6 allocation.")

base_map = dict(zip(
    base_alloc[base_layer_col].astype(str),
    base_alloc[base_fmt_col].astype(str)
))

# Normalize selected datatype labels.
for k, v in list(base_map.items()):
    if v == "E4M11_SUBNORMAL":
        base_map[k] = "E4M11_SN"

# Build full 53-layer candidate allocation: optimize measured layers,
# retain frozen Stage 5C.6 selections for unmeasured layers.
selected_map = dict(base_map)

# A complete evidence-supported 53-layer optimizer cannot be solved without
# inventing missing model-level risk. Instead, optimize only measured layers
# with a restricted MILP, holding the other 48 frozen.
opt_rows = joint_candidates[joint_candidates["layer"].isin(measured_layers)].copy()

if measured_layers:
    # Each layer x datatype is a binary decision variable.
    opt_rows["risk_pp"] = [
        measured_bitflip_risk(r.layer, r.format)
        for r in opt_rows.itertuples()
    ]
    opt_rows["objective"] = (
        opt_rows["quantization_loss_pp"] + LAMBDA_FAULT * opt_rows["risk_pp"]
    )

    n = len(opt_rows)
    constraints = []
    lb = []
    ub = []

    # One datatype per measured layer.
    for layer in measured_layers:
        row = np.zeros(n)
        row[np.where(opt_rows["layer"].values == layer)[0]] = 1
        constraints.append(row)
        lb.append(1)
        ub.append(1)

    # Storage budget for all 53 layers, including frozen selections.
    frozen_storage = 0
    frozen_loss = 0.0
    for layer in layers:
        if layer in measured_layers:
            continue
        fmt = base_map.get(layer, "FP32")
        frozen_storage += int(weight_counts[layer] * bit_width.get(fmt, 32) / 8)
        if fmt != "FP32":
            frozen_loss += q_lookup.get((layer, fmt), 0.0)

    row = opt_rows["storage_bytes"].to_numpy(dtype=float)
    constraints.append(row)
    lb.append(-np.inf)
    ub.append(storage_budget - frozen_storage)

    # Quantization-loss budget.
    row = opt_rows["quantization_loss_pp"].to_numpy(dtype=float)
    constraints.append(row)
    lb.append(-np.inf)
    ub.append(MAX_QUANTIZATION_LOSS_PP - frozen_loss)

    A = np.vstack(constraints)
    result = milp(
        c=opt_rows["objective"].to_numpy(dtype=float),
        integrality=np.ones(n),
        bounds=Bounds(np.zeros(n), np.ones(n)),
        constraints=LinearConstraint(A, np.asarray(lb), np.asarray(ub)),
        options={"time_limit": 120, "mip_rel_gap": 0.001}
    )

    if result.success and result.x is not None:
        chosen = opt_rows.iloc[np.where(result.x > 0.5)[0]]
        for r in chosen.itertuples():
            selected_map[r.layer] = r.format
        solver_status = "optimal_or_success"
        solver_message = str(result.message)
    else:
        solver_status = "no_feasible_measured_layer_solution"
        solver_message = str(result.message)
else:
    solver_status = "no_model_level_fault_layers"
    solver_message = "No measured model-level fault profiles available."

# ---------------------------------------------------------------------------
# 5. CREATE FINAL EVIDENCE-LIMITED ALLOCATION AND STORAGE REPORT
# ---------------------------------------------------------------------------

final_rows = []
for layer in layers:
    fmt = selected_map.get(layer, base_map.get(layer, "FP32"))
    nweights = weight_counts[layer]
    loss = 0.0 if fmt == "FP32" else q_lookup.get((layer, fmt), np.nan)

    final_rows.append({
        "layer": layer,
        "selected_datatype": fmt,
        "protection": "none",
        "weight_count": nweights,
        "storage_bytes_unprotected": int(np.ceil(nweights * bit_width.get(fmt, 32) / 8)),
        "isolated_quantization_loss_pp": loss,
        "model_level_fault_profile_available": layer in model_fault_layers,
        "fault_profile_scope": "measured_bitflip" if layer in model_fault_layers else "not_measured",
        "protection_benefit_claimed": False
    })

final_allocation = pd.DataFrame(final_rows)
final_allocation.to_csv(OUT / "stage5d_final_allocation.csv", index=False)

total_storage = int(final_allocation["storage_bytes_unprotected"].sum())
storage_reduction = 100 * (1 - total_storage / fp32_bytes)
total_quant_loss = float(final_allocation["isolated_quantization_loss_pp"].fillna(0).sum())

# ---------------------------------------------------------------------------
# 6. FINAL REPORT
# ---------------------------------------------------------------------------

datatype_distribution = (
    final_allocation["selected_datatype"].value_counts().to_dict()
)

report = {
    "experiment": "JFAMP_MobileNetV2_CIFAR100",
    "stage": "5D_consolidated_evidence_limited_joint_optimization",
    "timestamp_utc": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime()),
    "solver": {
        "method": "scipy.optimize.milp",
        "status": solver_status,
        "message": solver_message,
        "lambda_fault": LAMBDA_FAULT,
        "storage_budget_fraction": STORAGE_BUDGET_FRACTION,
        "quantization_loss_budget_pp": MAX_QUANTIZATION_LOSS_PP
    },
    "objective": "sum datatype quantization loss + lambda_fault * measured bit-flip residual risk",
    "original_optimizer_deviation": (
        "Random-corruption profile unavailable; measured bit-flip risk used for the "
        "measured subset only. No protection benefit credited without protection trials."
    ),
    "model": {
        "quantizable_layers": len(layers),
        "measured_model_fault_layers": len(model_fault_layers),
        "unmeasured_model_fault_layers": len(set(layers) - model_fault_layers),
        "payload_fault_layers": len(payload_fault_layers)
    },
    "clean_reference": {
        "fp32_accuracy_percent": clean_verification.get("fp32_accuracy_percent", 68.06),
        "mixed_candidate_accuracy_percent": clean_verification.get("candidate_accuracy_percent", 67.06),
        "mixed_candidate_delta_pp": clean_verification.get("delta_pp", -1.0),
        "mixed_candidate_storage_bytes": clean_verification.get("candidate_storage_bytes", None)
    },
    "final_allocation": {
        "datatype_distribution": datatype_distribution,
        "storage_bytes_unprotected": total_storage,
        "fp32_storage_bytes": fp32_bytes,
        "storage_reduction_percent": storage_reduction,
        "isolated_quantization_loss_proxy_pp": total_quant_loss,
        "protection_selected": "none",
        "protection_benefit_claimed": False
    },
    "fault_characterization": {
        "bitflip_model_level_trials_existing": int(len(fault_model)),
        "bitflip_payload_level_trials_existing": int(len(fault_payload)),
        "random_corruption_trials": 0,
        "protection_trials": 0,
        "fault_model_mix_assumption_original": {
            "bitflip": 0.5,
            "random_corruption": 0.5
        },
        "mix_applied_to_final_objective": False
    },
    "limitations": [
        "Only five layers have model-level fault measurements.",
        "The remaining 48 layers have payload-level diagnostics but no model-level accuracy fault profile.",
        "Random corruption was not experimentally characterized.",
        "Parity, SECDED and DMR were not validated on this MobileNetV2 payload implementation.",
        "No protection benefit is claimed.",
        "Storage estimates exclude metadata, alignment, activation storage and hardware controller overhead.",
        "Weight quantization is software serialization followed by FP32 reconstruction, not native low-precision hardware inference.",
        "This is not a complete 53-layer joint fault-and-protection optimum."
    ],
    "elapsed_seconds": time.time() - T0,
    "artifacts": {
        "final_allocation": str(OUT / "stage5d_final_allocation.csv"),
        "candidate_table": str(OUT / "stage5d_joint_candidate_table.csv"),
        "measured_fault_risk": str(OUT / "stage5d_measured_model_fault_risk.csv"),
        "payload_diagnostics": str(OUT / "stage5d_payload_fault_diagnostics.csv"),
        "report": str(OUT / "stage5d_final_report.json")
    }
}

with open(OUT / "stage5d_final_report.json", "w") as f:
    json.dump(report, f, indent=2)

# ---------------------------------------------------------------------------
# 7. DISPLAY FINAL RESULTS
# ---------------------------------------------------------------------------

print("\n" + "=" * 88)
print("STAGE 5D RESULTS")
print("=" * 88)
print("Solver status:", solver_status)
print("Solver message:", solver_message)
print("Layers:", len(final_allocation))
print("Model-level fault-measured layers:", len(model_fault_layers), "/ 53")
print("Datatype distribution:", datatype_distribution)
print(f"Estimated unprotected storage: {total_storage / 1024**2:.4f} MiB")
print(f"FP32 storage: {fp32_bytes / 1024**2:.4f} MiB")
print(f"Estimated storage reduction: {storage_reduction:.2f}%")
print(f"Sum isolated-layer quantization-loss proxy: {total_quant_loss:.4f} pp")
print("Protection benefit claimed: NO")
print("Random corruption characterized: NO")

print("\nFinal allocation:")
display(final_allocation)

print("\nSaved report:")
print(OUT / "stage5d_final_report.json")

print("\nElapsed time:", round(time.time() - T0, 2), "seconds")
print("=" * 88)
print("STAGE 5D CONSOLIDATED RUN COMPLETE")
print("=" * 88)
```

```python

# =============================================================================
# JFAMP — FINAL STAGE 5D CORRECTION AND END-TO-END VALIDATION
# MobileNetV2 / CIFAR-100
#
# Uses frozen, previously verified Stage 5C.6 allocation.
# Reconstructs and validates the complete mixed-precision model.
# Does not rerun completed fault experiments or invent protection benefits.
# =============================================================================

import os, json, time, copy, hashlib
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

T0 = time.time()

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
ALLOC = ROOT / "experiments/stage_05_allocation"
FINAL = ALLOC / "stage5d_final_verified"
FINAL.mkdir(parents=True, exist_ok=True)

print("=" * 88)
print("JFAMP — FINAL STAGE 5D VALIDATION")
print("=" * 88)

# ---------------------------------------------------------------------------
# 1. VERIFY REQUIRED RUNTIME OBJECTS AND FROZEN ARTIFACTS
# ---------------------------------------------------------------------------

required_objects = [
    "model", "val_loader", "encode_weights",
    "decode_weights", "evaluate_model"
]

for name in required_objects:
    if name not in globals():
        raise RuntimeError(f"Required runtime object missing: {name}")

candidate_path = ALLOC / "stage5c6_verified_candidate.pth"
allocation_path = ALLOC / "stage5c6_recovered_allocation.csv"
summary_path = ALLOC / "stage5c6_verification_summary.json"
reference_path = ALLOC / "stage5c6_verified_predictions.npz"
fp32_reference_path = ROOT / "experiments/stage_02_quantization/fp32_reference_predictions.npz"

for p in [candidate_path, allocation_path, summary_path, reference_path]:
    if not p.exists():
        raise FileNotFoundError(p)

allocation = pd.read_csv(allocation_path)

with open(summary_path) as f:
    stage5c6_summary = json.load(f)

print("Frozen allocation:", allocation.shape)
print("Frozen candidate checkpoint:", candidate_path.exists())

# ---------------------------------------------------------------------------
# 2. IDENTIFY ALLOCATION COLUMNS
# ---------------------------------------------------------------------------

def find_col(df, choices):
    normalized = {
        str(c).lower().replace("_", "").replace(" ", ""): c
        for c in df.columns
    }
    for choice in choices:
        key = choice.lower().replace("_", "").replace(" ", "")
        if key in normalized:
            return normalized[key]
    return None

layer_col = find_col(allocation, ["layer", "layer_name"])
format_col = find_col(allocation, ["selected_datatype", "format", "datatype"])

if layer_col is None or format_col is None:
    raise ValueError(f"Cannot identify allocation columns: {list(allocation.columns)}")

allocation[layer_col] = allocation[layer_col].astype(str)
allocation[format_col] = allocation[format_col].astype(str)
allocation[format_col] = allocation[format_col].replace({
    "E4M11_SUBNORMAL": "E4M11_SN"
})

# ---------------------------------------------------------------------------
# 3. LOAD THE ORIGINAL FP32 CHECKPOINT
# ---------------------------------------------------------------------------

checkpoint_path = ROOT / "checkpoints/mobilenetv2_cifar100_best.pth"

if not checkpoint_path.exists():
    raise FileNotFoundError(checkpoint_path)

checkpoint = torch.load(
    checkpoint_path,
    map_location="cpu",
    weights_only=False
)

if isinstance(checkpoint, dict):
    state = checkpoint.get("model_state_dict", checkpoint.get("state_dict", checkpoint))
else:
    raise ValueError("Unexpected checkpoint format.")

# Remove DataParallel prefix if present.
state = {
    k.replace("module.", "", 1) if k.startswith("module.") else k: v
    for k, v in state.items()
}

base_model = copy.deepcopy(model).cpu()
load_result = base_model.load_state_dict(state, strict=True)
base_model.eval()

# ---------------------------------------------------------------------------
# 4. APPLY THE FROZEN MIXED-PRECISION ALLOCATION
# ---------------------------------------------------------------------------

modules = dict(base_model.named_modules())

if len(allocation) != 53:
    raise ValueError(f"Expected 53 allocation rows, found {len(allocation)}")

actual_layer_names = {
    name for name, module in base_model.named_modules()
    if isinstance(module, (nn.Conv2d, nn.Linear))
}

if set(allocation[layer_col]) != actual_layer_names:
    missing = actual_layer_names - set(allocation[layer_col])
    extra = set(allocation[layer_col]) - actual_layer_names
    raise ValueError(f"Layer mismatch. Missing={missing}, extra={extra}")

format_payloads = {}
storage_rows = []

for row in allocation.itertuples(index=False):
    values = row._asdict()
    layer = str(values[layer_col])
    fmt = str(values[format_col])

    module = modules[layer]
    original_weight = module.weight.detach().cpu().numpy().copy()

    if fmt == "FP32":
        reconstructed = original_weight.copy()
        payload_bytes = original_weight.nbytes
        metadata = {"shape": list(original_weight.shape)}
    else:
        payload, metadata = encode_weights(original_weight, fmt)
        payload = np.asarray(payload).copy()
        reconstructed = decode_weights(
            payload.copy(), copy.deepcopy(metadata), fmt
        )
        reconstructed = np.asarray(reconstructed, dtype=np.float32).reshape(
            original_weight.shape
        )
        payload_bytes = int(payload.nbytes)

    if not np.isfinite(reconstructed).all():
        raise FloatingPointError(f"Non-finite reconstructed weights: {layer}/{fmt}")

    module.weight.data.copy_(
        torch.from_numpy(reconstructed.copy()).to(module.weight.dtype)
    )

    format_payloads[layer] = {
        "format": fmt,
        "payload_bytes": payload_bytes,
        "metadata": metadata
    }

    storage_rows.append({
        "layer": layer,
        "format": fmt,
        "weight_count": original_weight.size,
        "payload_bytes": payload_bytes,
        "fp32_bytes": original_weight.nbytes,
        "protection": "none"
    })

base_model.eval()

# ---------------------------------------------------------------------------
# 5. EVALUATE COMPLETE MODEL ON THE FULL VALIDATION SET
# ---------------------------------------------------------------------------

print("\nEvaluating final reconstructed model on validation set...")

final_eval = evaluate_model(base_model)

final_predictions = final_eval["predictions"]
final_accuracy = float(final_eval["accuracy"])
final_loss = float(final_eval["loss"])

# ---------------------------------------------------------------------------
# 6. LOAD REFERENCE PREDICTIONS
# ---------------------------------------------------------------------------

with np.load(reference_path) as data:
    candidate_reference_predictions = data["predictions"].copy()
    candidate_reference_labels = data["labels"].copy()

if len(final_predictions) != 5000:
    raise AssertionError("Final evaluation did not cover 5000 samples.")

if not np.array_equal(final_predictions.shape, candidate_reference_predictions.shape):
    raise AssertionError("Prediction shape mismatch.")

candidate_agreement = float(
    np.mean(final_predictions == candidate_reference_predictions) * 100
)

candidate_changed = int(
    np.sum(final_predictions != candidate_reference_predictions)
)

# Load FP32 reference predictions if available.
fp32_accuracy = float(stage5c6_summary.get("fp32_accuracy_percent", 68.06))

fp32_agreement = None
fp32_changed = None

if fp32_reference_path.exists():
    with np.load(fp32_reference_path) as data:
        fp32_predictions = data["predictions"].copy()

    fp32_agreement = float(
        np.mean(final_predictions == fp32_predictions) * 100
    )
    fp32_changed = int(
        np.sum(final_predictions != fp32_predictions)
    )

# ---------------------------------------------------------------------------
# 7. ACTUAL STORAGE AND CONSTRAINT CHECKS
# ---------------------------------------------------------------------------

storage_df = pd.DataFrame(storage_rows)

actual_storage = int(storage_df["payload_bytes"].sum())
fp32_storage = int(storage_df["fp32_bytes"].sum())

storage_reduction = 100.0 * (1 - actual_storage / fp32_storage)

accuracy_drop = fp32_accuracy - final_accuracy
accuracy_budget = 1.0
storage_budget_bytes = 0.60 * fp32_storage

accuracy_constraint_pass = accuracy_drop <= accuracy_budget + 1e-9
storage_constraint_pass = actual_storage <= storage_budget_bytes

# ---------------------------------------------------------------------------
# 8. PRESERVE ORIGINAL STAGE 4 FAULT COVERAGE LIMITATIONS
# ---------------------------------------------------------------------------

stage4d_path = ROOT / "experiments/stage_04_fault_injection/results/stage4d/stage4d_model_level_fault_trials.csv"
stage4c_path = ROOT / "experiments/stage_04_fault_injection/fault_trials/stage4c_full_layer/stage4c_full_layer_fault_characterization.csv"

stage4d = pd.read_csv(stage4d_path) if stage4d_path.exists() else pd.DataFrame()
stage4c = pd.read_csv(stage4c_path) if stage4c_path.exists() else pd.DataFrame()

model_fault_layers = (
    sorted(stage4d["layer"].astype(str).unique().tolist())
    if len(stage4d) else []
)

# ---------------------------------------------------------------------------
# 9. SAVE ALL FINAL ARTIFACTS
# ---------------------------------------------------------------------------

storage_df.to_csv(FINAL / "final_layerwise_storage.csv", index=False)

final_allocation = allocation.copy()
final_allocation["protection"] = "none"
final_allocation["protection_validated"] = False
final_allocation.to_csv(FINAL / "final_layerwise_allocation.csv", index=False)

np.savez_compressed(
    FINAL / "final_validation_predictions.npz",
    predictions=final_predictions,
    labels=candidate_reference_labels
)

torch.save(
    {
        "model_state_dict": base_model.state_dict(),
        "architecture": "MobileNetV2_CIFAR100",
        "allocation": final_allocation.to_dict(orient="records"),
        "validation_accuracy": final_accuracy,
        "validation_loss": final_loss
    },
    FINAL / "final_mixed_precision_model.pth"
)

report = {
    "experiment": "JFAMP_MobileNetV2_CIFAR100",
    "stage": "5D_final_verified_mixed_precision_candidate",
    "status": "completed",
    "validation_samples": 5000,
    "fp32_reference_accuracy_percent": fp32_accuracy,
    "final_mixed_precision_accuracy_percent": final_accuracy,
    "accuracy_drop_vs_fp32_pp": accuracy_drop,
    "final_validation_loss": final_loss,
    "stage5c6_prediction_agreement_percent": candidate_agreement,
    "changed_predictions_vs_stage5c6": candidate_changed,
    "fp32_prediction_agreement_percent": fp32_agreement,
    "changed_predictions_vs_fp32": fp32_changed,
    "actual_serialized_weight_storage_bytes": actual_storage,
    "fp32_weight_storage_bytes": fp32_storage,
    "storage_reduction_percent": storage_reduction,
    "storage_budget_fraction": 0.60,
    "accuracy_budget_pp": accuracy_budget,
    "accuracy_constraint_pass": bool(accuracy_constraint_pass),
    "storage_constraint_pass": bool(storage_constraint_pass),
    "datatype_distribution": final_allocation[format_col].value_counts().to_dict(),
    "fault_evidence": {
        "payload_level_trials": int(len(stage4c)),
        "model_level_trials": int(len(stage4d)),
        "model_level_measured_layers": model_fault_layers,
        "random_corruption_trials": 0
    },
    "protection": {
        "selected": "none",
        "parity_validated": False,
        "secded_validated": False,
        "dmr_validated": False,
        "protection_benefit_claimed": False
    },
    "methodology": {
        "weight_only_quantization": True,
        "inference_precision": "FP32 after decoding",
        "optimizer_output_used": "Frozen Stage 5C.6 verified allocation",
        "missing_fault_measurements_interpreted_as_zero": False
    },
    "limitations": [
        "This final validation uses the frozen Stage 5C.6 allocation.",
        "Only five layers have model-level bit-flip measurements.",
        "Random corruption was not characterized.",
        "Protection schemes were not experimentally validated for this model.",
        "No native low-precision hardware execution is claimed.",
        "No hardware area, timing, power or energy claims are made."
    ],
    "elapsed_seconds": time.time() - T0
}

with open(FINAL / "final_research_report.json", "w") as f:
    json.dump(report, f, indent=2)

# ---------------------------------------------------------------------------
# 10. FINAL SUMMARY
# ---------------------------------------------------------------------------

print("\n" + "=" * 88)
print("JFAMP — FINAL VALIDATION RESULTS")
print("=" * 88)

print(f"FP32 reference accuracy:       {fp32_accuracy:.4f}%")
print(f"Final mixed-precision accuracy: {final_accuracy:.4f}%")
print(f"Accuracy drop:                 {accuracy_drop:.4f} pp")
print(f"Validation loss:               {final_loss:.6f}")
print(f"Agreement with Stage 5C.6:     {candidate_agreement:.4f}%")
print(f"Changed predictions:           {candidate_changed}")
print(f"Actual weight storage:         {actual_storage / 1024**2:.4f} MiB")
print(f"FP32 weight storage:            {fp32_storage / 1024**2:.4f} MiB")
print(f"Storage reduction:              {storage_reduction:.4f}%")

print("\nConstraint checks:")
print("Accuracy <= 1.0 pp drop:", accuracy_constraint_pass)
print("Storage <= 60% FP32:", storage_constraint_pass)

print("\nFault coverage:")
print("Payload-level trials:", len(stage4c))
print("Model-level trials:", len(stage4d))
print("Model-level measured layers:", len(model_fault_layers))

print("\nProtection claims: NONE")
print("\nFinal artifacts:")
for p in sorted(FINAL.iterdir()):
    print(p.name, f"({p.stat().st_size / 1024:.1f} KiB)")

print("\nElapsed:", round(time.time() - T0, 2), "seconds")
print("=" * 88)
print("FINAL STAGE 5D VALIDATION COMPLETE")
print("=" * 88)
```

```python

# =============================================================================
# JFAMP — FINAL STAGE 6A
# ACTUAL PAYLOAD PROTECTION PROFILING
# MobileNetV2 / CIFAR-100
#
# 53 layers x 7 formats x 100 sampled words
# 2 fault scenarios x 4 protection strategies
#
# Stage 5 artifacts are read-only.
# =============================================================================

import numpy as np
import pandas as pd
import json
import time
from pathlib import Path

# --------------------------- CONFIGURATION ---------------------------

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
STAGE5 = ROOT / "experiments/stage_05_allocation"

OUT = ROOT / "experiments/stage_06_protection"
OUT.mkdir(parents=True, exist_ok=True)

TRIALS_PER_LAYER_FORMAT = 100
SEED = 20261114

FORMATS = [
    "FP16", "BF16", "E4M11_SN",
    "FP8_E4M3FN", "FP8_E5M2", "INT8", "INT4"
]

STRATEGIES = ["none", "parity", "secded", "dmr"]
SCENARIOS = ["single_bit", "double_bit"]

# --------------------------- INPUTS --------------------------------

assert "model" in globals()
assert "encode_weights" in globals()

characterization = pd.read_csv(
    STAGE5 / "stage5_unified_characterization.csv"
)

modules = dict(model.named_modules())

required_columns = {"layer", "format", "weight_count"}
assert required_columns.issubset(characterization.columns)

# One row per layer-format combination.
profile = characterization[
    characterization["format"].isin(FORMATS)
].copy()

profile = profile.drop_duplicates(["layer", "format"])

assert len(profile) == 53 * 7, (
    f"Expected 371 combinations, found {len(profile)}"
)

# --------------------------- ECC FUNCTIONS --------------------------

def make_secded(data):
    """
    Extended Hamming SECDED:
    32 data bits + 6 Hamming parity bits + 1 overall parity bit.
    Total = 39 bits.
    """
    code = np.zeros(39, dtype=np.uint8)

    parity_positions = {1, 2, 4, 8, 16, 32}
    data_positions = [
        p for p in range(1, 39)
        if p not in parity_positions
    ]

    assert len(data) == 32
    assert len(data_positions) == 32

    for bit, pos in zip(data, data_positions):
        code[pos - 1] = bit

    for parity_pos in parity_positions:
        value = 0
        for pos in range(1, 39):
            if pos & parity_pos:
                value ^= int(code[pos - 1])
        code[parity_pos - 1] = value

    code[38] = np.bitwise_xor.reduce(code[:38])
    return code


def decode_secded(code):
    """Return corrected data, detected, corrected, residual."""
    c = code.copy()

    parity_positions = {1, 2, 4, 8, 16, 32}
    data_positions = [
        p for p in range(1, 39)
        if p not in parity_positions
    ]

    syndrome = 0
    for parity_pos in parity_positions:
        value = 0
        for pos in range(1, 39):
            if pos & parity_pos:
                value ^= int(c[pos - 1])
        if value:
            syndrome |= parity_pos

    overall = int(np.bitwise_xor.reduce(c))

    detected = syndrome != 0 or overall != 0
    corrected = False

    if syndrome != 0 and overall == 1:
        if syndrome <= 38:
            c[syndrome - 1] ^= 1
            corrected = True

    elif syndrome == 0 and overall == 1:
        c[38] ^= 1
        corrected = True

    elif syndrome != 0 and overall == 0:
        # Double-bit error detected, not corrected.
        pass

    recovered = c[np.array(data_positions) - 1]
    residual = not np.array_equal(
        recovered,
        np.asarray(data, dtype=np.uint8)
    ) if False else False

    return recovered, detected, corrected


def protect_and_test(data, strategy, scenario, positions):
    """
    Inject faults into corresponding logical data-bit positions.
    For DMR, faults are applied to the first copy only.
    """

    data = np.asarray(data, dtype=np.uint8)
    corrupted_data = data.copy()

    for pos in positions:
        corrupted_data[pos] ^= 1

    if strategy == "none":
        recovered = corrupted_data
        detected = False
        corrected = False

    elif strategy == "parity":
        parity = int(np.bitwise_xor.reduce(data))
        corrupted_parity = parity

        # Data-bit corruption is injected, parity bit remains unchanged.
        parity_check = int(
            np.bitwise_xor.reduce(corrupted_data)
        ) ^ corrupted_parity

        detected = bool(parity_check)
        corrected = False
        recovered = corrupted_data

    elif strategy == "secded":
        code = make_secded(data)

        # Map logical data positions into the SECDED codeword.
        parity_positions = {1, 2, 4, 8, 16, 32}
        data_positions = [
            p for p in range(1, 39)
            if p not in parity_positions
        ]

        for pos in positions:
            code[data_positions[pos] - 1] ^= 1

        recovered, detected, corrected = decode_secded(code)

    elif strategy == "dmr":
        copy_a = data.copy()
        copy_b = data.copy()

        for pos in positions:
            copy_a[pos] ^= 1

        detected = not np.array_equal(copy_a, copy_b)
        corrected = False

        # DMR detects disagreement but cannot determine the correct copy.
        recovered = copy_a

    residual = not np.array_equal(recovered, data)

    return {
        "detected": bool(detected),
        "corrected": bool(corrected),
        "residual": bool(residual)
    }


# --------------------------- MAIN PROFILING -------------------------

rng = np.random.default_rng(SEED)

raw_records = []
storage_records = []

start_time = time.time()
combination_count = len(profile)

print("=" * 78)
print("JFAMP STAGE 6A — PROTECTION PROFILING")
print("=" * 78)
print("Layer-format combinations:", combination_count)
print("Sampled words per combination:", TRIALS_PER_LAYER_FORMAT)
print("Fault scenarios:", SCENARIOS)
print("Strategies:", STRATEGIES)
print()

for combo_idx, row in enumerate(
    profile.itertuples(index=False), start=1
):
    layer = row.layer
    fmt = row.format

    if layer not in modules:
        raise KeyError(f"Model layer not found: {layer}")

    weights = (
        modules[layer].weight.detach().cpu().float().numpy()
    )

    payload, metadata = encode_weights(weights, fmt)
    payload = np.asarray(payload, dtype=np.uint8).reshape(-1)

    bits = np.unpackbits(payload, bitorder="big")
    original_bits = len(bits)

    # Pad final partial 32-bit word with zeros.
    pad = (-original_bits) % 32
    if pad:
        bits = np.concatenate([
            bits,
            np.zeros(pad, dtype=np.uint8)
        ])

    words = bits.reshape(-1, 32)
    word_count = len(words)

    sampled_count = min(TRIALS_PER_LAYER_FORMAT, word_count)

    selected_words = rng.choice(
        word_count, size=sampled_count, replace=False
    )

    # Actual stored bits, including final byte padding.
    data_bytes = len(payload)

    storage_bits = {
        "none": original_bits,
        "parity": word_count * 33,
        "secded": word_count * 39,
        "dmr": word_count * 64
    }

    for strategy in STRATEGIES:
        storage_records.append({
            "layer": layer,
            "format": fmt,
            "strategy": strategy,
            "payload_bytes": data_bytes,
            "protected_bytes": int(
                np.ceil(storage_bits[strategy] / 8)
            ),
            "payload_bits": original_bits,
            "stored_bits": storage_bits[strategy],
            "overhead_percent": (
                100 * (
                    storage_bits[strategy] - original_bits
                ) / original_bits
            )
        })

    for local_trial, word_index in enumerate(selected_words):
        original_word = words[word_index].copy()

        for scenario in SCENARIOS:

            number_of_errors = (
                1 if scenario == "single_bit" else 2
            )

            positions = rng.choice(
                32,
                size=number_of_errors,
                replace=False
            ).tolist()

            for strategy in STRATEGIES:

                result = protect_and_test(
                    original_word,
                    strategy,
                    scenario,
                    positions
                )

                raw_records.append({
                    "layer": layer,
                    "format": fmt,
                    "trial": int(local_trial),
                    "word_index": int(word_index),
                    "scenario": scenario,
                    "strategy": strategy,
                    "fault_positions": str(positions),
                    **result
                })

    if combo_idx % 25 == 0 or combo_idx == combination_count:
        print(
            f"[{combo_idx}/{combination_count}] "
            f"{layer} | {fmt} | payload={data_bytes} bytes"
        )

# --------------------------- SAVE RAW DATA --------------------------

raw_df = pd.DataFrame(raw_records)
storage_df = pd.DataFrame(storage_records)

raw_path = OUT / "stage6a_protection_raw_trials.csv"
storage_path = OUT / "stage6a_protection_storage.csv"

raw_df.to_csv(raw_path, index=False)
storage_df.to_csv(storage_path, index=False)

# --------------------------- SUMMARIES ------------------------------

summary = (
    raw_df.groupby(
        ["format", "strategy", "scenario"],
        as_index=False
    )
    .agg(
        trials=("residual", "count"),
        detection_rate_percent=(
            "detected", lambda x: 100 * x.mean()
        ),
        correction_rate_percent=(
            "corrected", lambda x: 100 * x.mean()
        ),
        residual_error_rate_percent=(
            "residual", lambda x: 100 * x.mean()
        )
    )
)

layer_summary = (
    raw_df.groupby(
        ["layer", "format", "strategy", "scenario"],
        as_index=False
    )
    .agg(
        trials=("residual", "count"),
        detection_rate_percent=(
            "detected", lambda x: 100 * x.mean()
        ),
        correction_rate_percent=(
            "corrected", lambda x: 100 * x.mean()
        ),
        residual_error_rate_percent=(
            "residual", lambda x: 100 * x.mean()
        )
    )
)

summary_path = OUT / "stage6a_protection_summary.csv"
layer_summary_path = OUT / "stage6a_layer_protection_summary.csv"

summary.to_csv(summary_path, index=False)
layer_summary.to_csv(layer_summary_path, index=False)

# --------------------------- REPORT ---------------------------------

report = {
    "experiment": "JFAMP MobileNetV2 CIFAR100 Stage 6A",
    "status": "completed",
    "layers": int(profile["layer"].nunique()),
    "formats": FORMATS,
    "layer_format_combinations": int(len(profile)),
    "sampled_words_per_combination": TRIALS_PER_LAYER_FORMAT,
    "strategies": STRATEGIES,
    "scenarios": SCENARIOS,
    "raw_trial_count": int(len(raw_df)),
    "fault_model": (
        "Faults injected into logical data-bit positions of sampled "
        "32-bit payload words. Parity/check-bit corruption excluded."
    ),
    "codeword_layout": {
        "data_word_bits": 32,
        "parity_bits": 1,
        "secded_bits": 7,
        "dmr_copies": 2
    },
    "limitations": [
        "Codeword-level software profiling; not hardware reliability testing.",
        "Faults are injected into data bits, not parity/check bits.",
        "DMR detects mismatch but does not correct it.",
        "Layer-level network accuracy under protected faults is not measured here.",
        "Storage figures include codeword packing and final-byte padding, "
        "but exclude metadata, alignment and controller overhead."
    ],
    "elapsed_seconds": round(time.time() - start_time, 3),
    "outputs": {
        "raw_trials": str(raw_path),
        "summary": str(summary_path),
        "layer_summary": str(layer_summary_path),
        "storage": str(storage_path)
    }
}

report_path = OUT / "stage6a_protection_report.json"

with open(report_path, "w") as f:
    json.dump(report, f, indent=2)

# --------------------------- DISPLAY --------------------------------

print("\n" + "=" * 78)
print("STAGE 6A — PROTECTION PROFILE SUMMARY")
print("=" * 78)

print("\nTOTAL TRIALS:", len(raw_df))
print("ELAPSED:", round(time.time() - start_time, 2), "seconds")

print("\nMECHANISM-LEVEL RESULTS")
display(
    summary.round(3)
)

print("\nAVERAGE STORAGE OVERHEAD")
display(
    storage_df.groupby("strategy", as_index=False)
    .agg(
        average_overhead_percent=("overhead_percent", "mean"),
        total_payload_bytes=("payload_bytes", "sum"),
        total_protected_bytes=("protected_bytes", "sum")
    ).round(3)
)

print("\nSAVED ARTIFACTS")
for p in [raw_path, summary_path, layer_summary_path,
          storage_path, report_path]:
    print(p)

print("\nSTAGE 6A PROFILING COMPLETE")
print("=" * 78)
```

```python

# =============================================================================
# JFAMP — FINAL STAGE 6B
# JOINT DATATYPE–PROTECTION CANDIDATE GENERATION
# =============================================================================

import pandas as pd
import numpy as np
import json
from pathlib import Path

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
STAGE5 = ROOT / "experiments/stage_05_allocation"
STAGE6 = ROOT / "experiments/stage_06_protection"

OUT = STAGE6 / "stage6b_candidates"
OUT.mkdir(parents=True, exist_ok=True)

# 1. Load existing measured data.
quant = pd.read_csv(
    STAGE5 / "stage5_unified_characterization.csv"
)

storage = pd.read_csv(
    STAGE6 / "stage6a_protection_storage.csv"
)

trials = pd.read_csv(
    STAGE6 / "stage6a_protection_raw_trials.csv"
)

# 2. Build quantization cost Q.
quant = quant[
    quant["format"].isin([
        "FP16", "BF16", "E4M11_SN",
        "FP8_E4M3FN", "FP8_E5M2", "INT8", "INT4"
    ])
].copy()

quant = quant.drop_duplicates(["layer", "format"])

quant["Q_quantization_loss_pp"] = (
    quant["clean_accuracy_delta_pp"].clip(lower=0)
)

# 3. Calculate residual risk separately for each fault scenario.
scenario_risk = (
    trials.groupby(
        ["layer", "format", "strategy", "scenario"],
        as_index=False
    )
    .agg(
        residual_error_rate_percent=(
            "residual", lambda x: 100 * x.mean()
        ),
        detection_rate_percent=(
            "detected", lambda x: 100 * x.mean()
        ),
        correction_rate_percent=(
            "corrected", lambda x: 100 * x.mean()
        ),
        trial_count=("residual", "count")
    )
)

# Equal weighting of single-bit and double-bit scenarios.
risk = (
    scenario_risk.groupby(
        ["layer", "format", "strategy"],
        as_index=False
    )
    .agg(
        R_residual_fault_risk_percent=(
            "residual_error_rate_percent", "mean"
        ),
        detection_rate_percent=("detection_rate_percent", "mean"),
        correction_rate_percent=("correction_rate_percent", "mean"),
        total_trials=("trial_count", "sum")
    )
)

# 4. Merge actual codeword storage measurements.
storage = storage.rename(columns={
    "protected_bytes": "S_protected_bytes",
    "payload_bytes": "S_payload_bytes"
})

storage = storage[
    [
        "layer", "format", "strategy",
        "S_payload_bytes", "S_protected_bytes",
        "stored_bits", "overhead_percent"
    ]
]

candidates = (
    quant.merge(
        risk,
        on=["layer", "format"],
        how="inner"
    )
    .merge(
        storage,
        on=["layer", "format", "strategy"],
        how="inner"
    )
)

# 5. Define the combined objective cost.
LAMBDA_FAULT = 1.0

candidates["lambda_fault"] = LAMBDA_FAULT

candidates["joint_objective_cost"] = (
    candidates["Q_quantization_loss_pp"]
    + LAMBDA_FAULT *
    candidates["R_residual_fault_risk_percent"]
)

# 6. Establish storage and accuracy budgets.
FP32_WEIGHT_BYTES = int(
    quant.drop_duplicates("layer")["weight_count"].sum() * 4
)

STORAGE_BUDGET_FRACTION = 0.60
STORAGE_BUDGET_BYTES = (
    FP32_WEIGHT_BYTES * STORAGE_BUDGET_FRACTION
)

MAX_QUANTIZATION_LOSS_PP = 1.0

# 7. Candidate-level eligibility.
candidates["quantization_budget_eligible"] = (
    candidates["Q_quantization_loss_pp"]
    <= MAX_QUANTIZATION_LOSS_PP
)

candidates["storage_individually_eligible"] = (
    candidates["S_protected_bytes"]
    <= STORAGE_BUDGET_BYTES
)

# 8. Sort and save.
candidates = candidates.sort_values(
    ["layer", "joint_objective_cost"]
).reset_index(drop=True)

candidate_path = OUT / "stage6b_joint_candidates.csv"
scenario_path = OUT / "stage6b_scenario_risk.csv"

candidates.to_csv(candidate_path, index=False)
scenario_risk.to_csv(scenario_path, index=False)

# 9. Save experiment configuration.
report = {
    "stage": "6B",
    "experiment": "JFAMP MobileNetV2 CIFAR100",
    "candidate_count": int(len(candidates)),
    "layer_count": int(candidates["layer"].nunique()),
    "datatype_count": int(candidates["format"].nunique()),
    "protection_count": int(candidates["strategy"].nunique()),
    "lambda_fault": LAMBDA_FAULT,
    "fault_scenario_weights": {
        "single_bit": 0.5,
        "double_bit": 0.5
    },
    "storage_budget_fraction": STORAGE_BUDGET_FRACTION,
    "fp32_weight_storage_bytes": FP32_WEIGHT_BYTES,
    "storage_budget_bytes": STORAGE_BUDGET_BYTES,
    "quantization_loss_budget_pp": MAX_QUANTIZATION_LOSS_PP,
    "quantization_cost": (
        "max(clean_accuracy_delta_pp, 0)"
    ),
    "fault_risk_cost": (
        "Mean measured residual error rate across "
        "single-bit and double-bit scenarios"
    ),
    "objective": "Q + lambda_fault * R",
    "outputs": {
        "candidates": str(candidate_path),
        "scenario_risk": str(scenario_path)
    }
}

with open(OUT / "stage6b_candidate_report.json", "w") as f:
    json.dump(report, f, indent=2)

# 10. Display summary.
print("=" * 78)
print("JFAMP — STAGE 6B CANDIDATE GENERATION")
print("=" * 78)

print("Candidate rows:", len(candidates))
print("Unique layers:", candidates["layer"].nunique())
print("Formats:", sorted(candidates["format"].unique()))
print("Protection strategies:", sorted(candidates["strategy"].unique()))

print("\nCANDIDATE DISTRIBUTION")
display(
    candidates.groupby(
        ["format", "strategy"], as_index=False
    ).agg(
        candidates=("layer", "count"),
        average_Q=("Q_quantization_loss_pp", "mean"),
        average_R=("R_residual_fault_risk_percent", "mean"),
        average_storage_bytes=("S_protected_bytes", "mean")
    ).round(4)
)

print("\nSTORAGE CONSTRAINT")
print("FP32 weight bytes:", FP32_WEIGHT_BYTES)
print("60% storage budget:", STORAGE_BUDGET_BYTES)

print("\nOBJECTIVE EXAMPLES")
display(
    candidates[
        [
            "layer", "format", "strategy",
            "Q_quantization_loss_pp",
            "R_residual_fault_risk_percent",
            "S_protected_bytes",
            "joint_objective_cost"
        ]
    ].head(12).round(5)
)

print("\nSaved:", candidate_path)
print("Saved:", scenario_path)
print("STAGE 6B COMPLETE")
print("=" * 78)
```

```python

# =============================================================================
# JFAMP — FINAL STAGE 6C
# CONSTRAINED JOINT DATATYPE–PROTECTION OPTIMIZATION
# =============================================================================

import json
import numpy as np
import pandas as pd

from pathlib import Path
from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import lil_matrix

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
STAGE6 = ROOT / "experiments/stage_06_protection"
IN_DIR = STAGE6 / "stage6b_candidates"
OUT = STAGE6 / "stage6c_optimization"
OUT.mkdir(parents=True, exist_ok=True)

# -------------------------------------------------------------------------
# 1. Load candidates
# -------------------------------------------------------------------------

df = pd.read_csv(IN_DIR / "stage6b_joint_candidates.csv")

required = [
    "layer", "format", "strategy",
    "Q_quantization_loss_pp",
    "R_residual_fault_risk_percent",
    "S_protected_bytes"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

df = df.dropna(subset=required).copy()
df = df.reset_index(drop=True)

df["Q"] = df["Q_quantization_loss_pp"].clip(lower=0)
df["R_fraction"] = (
    df["R_residual_fault_risk_percent"] / 100.0
)
df["S"] = df["S_protected_bytes"]

if not np.isfinite(df[["Q", "R_fraction", "S"]].to_numpy()).all():
    raise ValueError("Non-finite candidate values found.")

# -------------------------------------------------------------------------
# 2. Optimization parameters
# -------------------------------------------------------------------------

LAMBDA_FAULT = 1.0
STORAGE_BUDGET_FRACTION = 0.60
QUANTIZATION_BUDGET_PP = 1.0

# Same FP32 reference storage used in Stage 6B.
FP32_WEIGHT_BYTES = 9_271_040
STORAGE_BUDGET_BYTES = (
    FP32_WEIGHT_BYTES * STORAGE_BUDGET_FRACTION
)

# Objective:
# Q is in percentage points.
# R_fraction is residual-error probability in [0,1].
# lambda_fault = 1 means one full unit of residual probability
# is assigned the same numerical weight as one percentage point.
df["objective_cost"] = (
    df["Q"] + LAMBDA_FAULT * df["R_fraction"]
)

layers = sorted(df["layer"].unique())
n_layers = len(layers)
n_candidates = len(df)

if n_layers != 53:
    raise ValueError(
        f"Expected 53 layers, found {n_layers}."
    )

# Every layer must have at least one candidate.
candidate_counts = df.groupby("layer").size()
if not (candidate_counts > 0).all():
    raise ValueError("A layer has no eligible candidates.")

# -------------------------------------------------------------------------
# 3. Build MILP constraints
# -------------------------------------------------------------------------

# Decision variable x_i is binary:
# 1 = select candidate i, 0 = do not select it.

A = lil_matrix((n_layers + 2, n_candidates), dtype=float)
lower = np.full(n_layers + 2, -np.inf)
upper = np.full(n_layers + 2, np.inf)

# Exactly one datatype-protection choice per layer.
for row_idx, layer in enumerate(layers):
    indices = np.where(df["layer"].to_numpy() == layer)[0]
    A[row_idx, indices] = 1.0
    lower[row_idx] = 1.0
    upper[row_idx] = 1.0

# Global storage constraint.
A[n_layers, :] = df["S"].to_numpy()
lower[n_layers] = -np.inf
upper[n_layers] = STORAGE_BUDGET_BYTES

# Additive quantization-loss screening constraint.
A[n_layers + 1, :] = df["Q"].to_numpy()
lower[n_layers + 1] = -np.inf
upper[n_layers + 1] = QUANTIZATION_BUDGET_PP

constraints = LinearConstraint(
    A.tocsr(),
    lower,
    upper
)

# -------------------------------------------------------------------------
# 4. Solve
# -------------------------------------------------------------------------

result = milp(
    c=df["objective_cost"].to_numpy(),
    integrality=np.ones(n_candidates),
    bounds=Bounds(
        np.zeros(n_candidates),
        np.ones(n_candidates)
    ),
    constraints=constraints,
    options={"time_limit": 300}
)

if not result.success:
    raise RuntimeError(
        "Joint optimization did not produce an optimal solution. "
        f"Status={result.status}; message={result.message}"
    )

# -------------------------------------------------------------------------
# 5. Recover selected allocation
# -------------------------------------------------------------------------

df["selected"] = result.x > 0.5
selected = df[df["selected"]].copy()

if len(selected) != n_layers:
    raise RuntimeError(
        f"Expected {n_layers} selections, got {len(selected)}."
    )

selected = selected.sort_values("layer").reset_index(drop=True)

total_storage = float(selected["S"].sum())
total_Q = float(selected["Q"].sum())
mean_R = float(selected["R_fraction"].mean())
total_objective = float(selected["objective_cost"].sum())

if total_storage > STORAGE_BUDGET_BYTES + 1e-5:
    raise RuntimeError("Storage constraint violated.")

if total_Q > QUANTIZATION_BUDGET_PP + 1e-6:
    raise RuntimeError("Quantization constraint violated.")

# -------------------------------------------------------------------------
# 6. Save allocation and solver evidence
# -------------------------------------------------------------------------

allocation_columns = [
    "layer", "format", "strategy",
    "Q_quantization_loss_pp",
    "R_residual_fault_risk_percent",
    "R_fraction",
    "S_payload_bytes",
    "S_protected_bytes",
    "overhead_percent",
    "objective_cost"
]

allocation_path = OUT / "stage6c_joint_allocation.csv"
selected[allocation_columns].to_csv(
    allocation_path, index=False
)

format_counts = (
    selected["format"].value_counts().to_dict()
)

protection_counts = (
    selected["strategy"].value_counts().to_dict()
)

summary = {
    "stage": "6C",
    "status": "optimal",
    "solver": "scipy.optimize.milp",
    "solver_message": str(result.message),
    "selected_layer_count": int(len(selected)),
    "candidate_count": int(n_candidates),
    "lambda_fault": LAMBDA_FAULT,
    "objective": "sum(Q + lambda_fault * R_fraction)",
    "risk_definition": (
        "Mean measured residual data-bit error rate across "
        "single-bit and double-bit scenarios, equally weighted"
    ),
    "storage_budget_bytes": STORAGE_BUDGET_BYTES,
    "selected_storage_bytes": total_storage,
    "storage_budget_utilization_percent": (
        100 * total_storage / STORAGE_BUDGET_BYTES
    ),
    "fp32_storage_bytes": FP32_WEIGHT_BYTES,
    "storage_reduction_vs_fp32_percent": (
        100 * (1 - total_storage / FP32_WEIGHT_BYTES)
    ),
    "quantization_loss_budget_pp": QUANTIZATION_BUDGET_PP,
    "selected_additive_quantization_loss_pp": total_Q,
    "mean_residual_risk_fraction": mean_R,
    "total_objective": total_objective,
    "format_distribution": format_counts,
    "protection_distribution": protection_counts,
    "allocation_file": str(allocation_path),
    "limitations": [
        "Residual risk is a codeword-level proxy, not end-to-end model accuracy.",
        "Fault trials inject errors into data bits, not protection bits.",
        "Random-corruption scenarios are not included.",
        "Quantization loss is an additive isolated-layer screening proxy.",
        "Quantizer metadata and implementation/controller overhead are excluded.",
        "No FP32 or protected-FP32 candidates were profiled in Stage 6A."
    ]
}

summary_path = OUT / "stage6c_optimization_summary.json"
with open(summary_path, "w") as f:
    json.dump(summary, f, indent=2)

# -------------------------------------------------------------------------
# 7. Display final optimization result
# -------------------------------------------------------------------------

print("=" * 78)
print("JFAMP — STAGE 6C JOINT OPTIMIZATION")
print("=" * 78)

print("\nSOLVER:", result.message)
print("Selected layers:", len(selected))
print("Candidate decisions:", n_candidates)

print("\nCONSTRAINT RESULTS")
print(f"Storage: {total_storage:,.0f} / {STORAGE_BUDGET_BYTES:,.0f} bytes")
print(f"Storage budget used: {100*total_storage/STORAGE_BUDGET_BYTES:.2f}%")
print(f"Reduction vs FP32: {100*(1-total_storage/FP32_WEIGHT_BYTES):.2f}%")
print(f"Additive quantization loss: {total_Q:.5f} / {QUANTIZATION_BUDGET_PP:.2f} pp")
print(f"Mean residual risk: {100*mean_R:.2f}%")
print(f"Objective value: {total_objective:.5f}")

print("\nDATATYPE ALLOCATION")
display(
    selected["format"].value_counts()
    .rename_axis("datatype")
    .reset_index(name="layers")
)

print("\nPROTECTION ALLOCATION")
display(
    selected["strategy"].value_counts()
    .rename_axis("protection")
    .reset_index(name="layers")
)

print("\nJOINT ALLOCATION — FIRST 20 LAYERS")
display(
    selected[allocation_columns].head(20).round(5)
)

print("\nSaved allocation:", allocation_path)
print("Saved summary:", summary_path)

print("\nSTAGE 6C OPTIMIZATION COMPLETE")
print("=" * 78)
```

```python

# =============================================================================
# JFAMP — STAGE 6D (FINAL)
# RECONSTRUCT, VALIDATE AND PACKAGE THE JOINT ALLOCATION
# =============================================================================

import copy
import json
import numpy as np
import pandas as pd
import torch

from pathlib import Path

ROOT = Path("/kaggle/working/JFAMP_MobileNetV2")
STAGE5 = ROOT / "experiments/stage_05_allocation"
STAGE6 = ROOT / "experiments/stage_06_protection"

ALLOC_PATH = (
    STAGE6 / "stage6c_optimization" /
    "stage6c_joint_allocation.csv"
)

OUT = STAGE6 / "stage6d_final"
OUT.mkdir(parents=True, exist_ok=True)

# -------------------------------------------------------------------------
# 1. Load and validate the selected allocation
# -------------------------------------------------------------------------

allocation = pd.read_csv(ALLOC_PATH)

assert len(allocation) == 53, (
    f"Expected 53 layer assignments, found {len(allocation)}"
)

assert allocation["layer"].is_unique, (
    "Duplicate layer assignments detected."
)

modules = dict(model.named_modules())

missing_layers = [
    name for name in allocation["layer"]
    if name not in modules
]

if missing_layers:
    raise RuntimeError(
        f"Allocation contains unknown model layers: {missing_layers}"
    )

# -------------------------------------------------------------------------
# 2. Reconstruct the selected mixed-precision model
# -------------------------------------------------------------------------

final_model = copy.deepcopy(model)
final_modules = dict(final_model.named_modules())

metadata_records = []
payload_records = []

for row in allocation.itertuples(index=False):

    layer_name = row.layer
    fmt = row.format

    module = final_modules[layer_name]

    if not hasattr(module, "weight") or module.weight is None:
        raise RuntimeError(
            f"Layer {layer_name} has no quantizable weight."
        )

    original_weight = (
        module.weight.detach().cpu().numpy().astype(np.float32)
    )

    # Encode using the actual Stage 5 codec.
    payload, metadata = encode_weights(original_weight, fmt)

    # Decode the serialized payload back to model weights.
    decoded = decode_weights(payload, metadata, fmt)

    if isinstance(decoded, torch.Tensor):
        decoded_tensor = decoded.detach().cpu().to(
            dtype=module.weight.dtype
        )
    else:
        decoded_tensor = torch.as_tensor(
            np.asarray(decoded),
            dtype=module.weight.dtype
        )

    if tuple(decoded_tensor.shape) != tuple(module.weight.shape):
        raise RuntimeError(
            f"Shape mismatch for {layer_name}: "
            f"{tuple(decoded_tensor.shape)} vs "
            f"{tuple(module.weight.shape)}"
        )

    if not torch.isfinite(decoded_tensor).all():
        raise RuntimeError(
            f"Non-finite reconstructed weights in {layer_name}"
        )

    with torch.no_grad():
        module.weight.copy_(
            decoded_tensor.to(module.weight.device)
        )

    # Measure actual serialized payload bytes.
    if isinstance(payload, torch.Tensor):
        payload_array = payload.detach().cpu().numpy()
    else:
        payload_array = np.asarray(payload)

    payload_bytes = int(payload_array.nbytes)

    # Measure quantizer metadata separately.
    metadata_json = json.dumps(
        metadata,
        default=lambda x: (
            x.item() if isinstance(x, np.generic)
            else x.tolist() if isinstance(x, np.ndarray)
            else str(x)
        ),
        sort_keys=True
    )

    metadata_bytes = len(metadata_json.encode("utf-8"))

    payload_records.append({
        "layer": layer_name,
        "format": fmt,
        "payload_bytes": payload_bytes,
        "metadata_bytes": metadata_bytes,
        "total_serialized_bytes": (
            payload_bytes + metadata_bytes
        )
    })

    metadata_records.append({
        "layer": layer_name,
        "format": fmt,
        "codec_metadata": metadata_json
    })

# -------------------------------------------------------------------------
# 3. Evaluate FP32 baseline and reconstructed joint model
# -------------------------------------------------------------------------

print("=" * 78)
print("JFAMP — STAGE 6D FINAL VALIDATION")
print("=" * 78)

print("\nEvaluating FP32 baseline...")
baseline_result = evaluate(base_model)

print("\nEvaluating reconstructed joint allocation...")
joint_result = evaluate(final_model)

print("\nFP32 evaluation result:")
print(baseline_result)

print("\nJoint allocation evaluation result:")
print(joint_result)

# -------------------------------------------------------------------------
# 4. Calculate storage accounting
# -------------------------------------------------------------------------

storage_df = pd.DataFrame(payload_records)

total_payload_bytes = int(storage_df["payload_bytes"].sum())
total_metadata_bytes = int(storage_df["metadata_bytes"].sum())
total_serialized_bytes = (
    total_payload_bytes + total_metadata_bytes
)

fp32_weight_bytes = 9_271_040

payload_reduction = (
    100 * (1 - total_payload_bytes / fp32_weight_bytes)
)

serialized_reduction = (
    100 * (1 - total_serialized_bytes / fp32_weight_bytes)
)

# Optimizer's SECDED storage estimate.
optimizer_protected_bytes = int(
    allocation["S_protected_bytes"].sum()
)

# -------------------------------------------------------------------------
# 5. Save final model and allocation
# -------------------------------------------------------------------------

model_path = OUT / "jfamp_mobilenetv2_joint_final.pth"
allocation_path = OUT / "jfamp_final_layerwise_allocation.csv"
storage_path = OUT / "jfamp_final_storage_accounting.csv"
metadata_path = OUT / "jfamp_final_codec_metadata.csv"

torch.save(final_model.state_dict(), model_path)

allocation.to_csv(allocation_path, index=False)
storage_df.to_csv(storage_path, index=False)

pd.DataFrame(metadata_records).to_csv(
    metadata_path, index=False
)

# Save complete evaluation and experiment report.
report = {
    "project": "JFAMP — MobileNetV2 CIFAR-100",
    "stage": "6D — Final",
    "selected_layers": int(len(allocation)),
    "datatype_distribution": {
        str(k): int(v)
        for k, v in allocation["format"].value_counts().items()
    },
    "protection_distribution": {
        str(k): int(v)
        for k, v in allocation["strategy"].value_counts().items()
    },
    "validation": {
        "fp32_evaluation_result": str(baseline_result),
        "joint_evaluation_result": str(joint_result)
    },
    "storage": {
        "fp32_reference_weight_bytes": fp32_weight_bytes,
        "actual_payload_bytes": total_payload_bytes,
        "codec_metadata_json_bytes": total_metadata_bytes,
        "payload_plus_metadata_bytes": total_serialized_bytes,
        "payload_reduction_percent": payload_reduction,
        "payload_plus_metadata_reduction_percent": serialized_reduction,
        "optimizer_secded_codeword_storage_bytes": optimizer_protected_bytes
    },
    "optimizer_constraints": {
        "storage_budget_fraction": 0.60,
        "quantization_loss_budget_pp": 1.0,
        "selected_additive_quantization_loss_pp": float(
            allocation["Q_quantization_loss_pp"].sum()
        ),
        "optimizer_objective": float(
            allocation["objective_cost"].sum()
        )
    },
    "limitations": [
        "The reconstructed model validates datatype allocation using the existing codecs.",
        "This is weight-only quantization; biases, activations and other model state remain unchanged.",
        "Codec metadata is measured separately and is not included in the Stage 6C MILP storage constraint.",
        "SECDED storage is an estimated codeword overhead; SECDED is not implemented in this model inference path.",
        "No end-to-end fault-injected network evaluation was performed for the selected joint allocation.",
        "The protection profile covers single-bit and double-bit data-bit injections, not check-bit faults.",
        "Random-corruption fault scenarios were not measured.",
        "The additive quantization-loss constraint is a screening proxy; final validation accuracy is the authoritative clean-model measurement."
    ],
    "artifacts": {
        "model": str(model_path),
        "allocation": str(allocation_path),
        "storage_accounting": str(storage_path),
        "codec_metadata": str(metadata_path)
    }
}

report_path = OUT / "jfamp_final_research_report.json"

with open(report_path, "w") as f:
    json.dump(report, f, indent=2)

# -------------------------------------------------------------------------
# 6. Final summary
# -------------------------------------------------------------------------

print("\n" + "=" * 78)
print("FINAL JFAMP ARTIFACT SUMMARY")
print("=" * 78)

print("Layer assignments:", len(allocation))

print("\nDatatype distribution:")
print(allocation["format"].value_counts())

print("\nProtection distribution:")
print(allocation["strategy"].value_counts())

print("\nStorage accounting:")
print(f"FP32 reference:              {fp32_weight_bytes:,} bytes")
print(f"Actual encoded payload:      {total_payload_bytes:,} bytes")
print(f"Codec metadata (JSON):       {total_metadata_bytes:,} bytes")
print(f"Payload + metadata:          {total_serialized_bytes:,} bytes")
print(f"Optimizer SECDED estimate:   {optimizer_protected_bytes:,} bytes")
print(f"Payload reduction:           {payload_reduction:.2f}%")
print(f"Payload + metadata reduction:{serialized_reduction:.2f}%")

print("\nFINAL FILES")
print(model_path)
print(allocation_path)
print(storage_path)
print(metadata_path)
print(report_path)

print("\nSTAGE 6D COMPLETE — FINAL PROJECT STAGE")
print("No Stage 7.")
print("=" * 78)
```
